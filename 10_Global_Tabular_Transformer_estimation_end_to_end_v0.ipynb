### Building features

```python
def haversine_distance(lat1, lon1, lat2, lon2, radius_km=6371.0):
    arrays = (np.radians(np.asarray(value, dtype="float64")) for value in (lat1, lon1, lat2, lon2))
    lat1, lon1, lat2, lon2 = arrays
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat / 2.0) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2.0) ** 2
    c = 2.0 * np.arctan2(np.sqrt(a), np.sqrt(1.0 - a))
    return radius_km * c


def parse_storm_datetime(series):
    parsed = pd.to_datetime(series, format="%d-%b-%y %H:%M:%S", errors="coerce")
    future_century = parsed.dt.year > 2025
    return parsed.where(~future_century, parsed - pd.DateOffset(years=100))


MONTH_NAME_TO_NUMBER = {
    "January": 1, "February": 2, "March": 3, "April": 4,
    "May": 5, "June": 6, "July": 7, "August": 8,
    "September": 9, "October": 10, "November": 11, "December": 12,
}


def month_cyclical(month_name, fallback_month):
    """sin / cos of MONTH_NAME on the 12-month circle.
    """
    number = (
        month_name.astype("string")
        .str.strip()
        .str.capitalize()
        .map(MONTH_NAME_TO_NUMBER)
    )
    number = number.astype("Float64").fillna(fallback_month).astype("float64")
    angle = 2.0 * np.pi * (number - 1.0) / 12.0
    return np.sin(angle), np.cos(angle)


def build_base_event_frame(raw):
    out = raw.copy()
    out["BEGIN_DATE_TIME"] = parse_storm_datetime(out["BEGIN_DATE_TIME"])
    out["END_DATE_TIME"] = parse_storm_datetime(out["END_DATE_TIME"])
    out = out.loc[out["BEGIN_DATE_TIME"].notna()].copy()

    out.index = pd.DatetimeIndex(out["BEGIN_DATE_TIME"], name="Date")
    out = out.sort_index(kind="stable")

    begin = out["BEGIN_DATE_TIME"]
    end = out["END_DATE_TIME"]
    out["YEAR_BEGIN"] = begin.dt.year.astype("int64")
    out["MONTH_BEGIN"] = begin.dt.month.astype("int64")
    out["DAY_BEGIN"] = begin.dt.day.astype("int64")
    out["HOUR_BEGIN"] = begin.dt.hour.astype("int64")
    out["YEAR_END"] = end.dt.year.astype("Int64")
    out["MONTH_END"] = end.dt.month.astype("Int64")
    out["DAY_END"] = end.dt.day.astype("Int64")
    out["HOUR_END"] = end.dt.hour.astype("Int64")
    out["DAY_DURATION"] = (end - begin).dt.total_seconds() / 86400.0
    out["HOUR_DURATION"] = (end - begin).dt.total_seconds() / 3600.0

    out["MONTH_SIN"], out["MONTH_COS"] = month_cyclical(out["MONTH_NAME"], out["MONTH_BEGIN"])

    out['DISTANCE'] = haversine_distance(out["BEGIN_LAT"], out["BEGIN_LON"], out["END_LAT"], out["END_LON"])
    out["EXPOSURE"] = 1.0

    # prediction targets: direct + indirect
    out["INJURIES"] = out["INJURIES_DIRECT"] + out["INJURIES_INDIRECT"]
    out["DEATHS"] = out["DEATHS_DIRECT"] + out["DEATHS_INDIRECT"]
    out["CASUALTY"] = out["INJURIES"] + out["DEATHS"]


    # drop columns
    drop_replaced_features = [
        "BEGIN_LOCATION", "END_LOCATION", "EVENT_ID",
        "BEGIN_DATE_TIME", "END_DATE_TIME",
        "EPISODE_NARRATIVE", "EVENT_NARRATIVE",
        "BEGIN_LAT", "BEGIN_LON", "END_LAT", "END_LON",
        "DAY_BEGIN","DAY_END","HOUR_BEGIN","HOUR_END",
        "MAGNITUDE", "MAGNITUDE_TYPE", "TOR_F_SCALE", "TOR_WIDTH", "TOR_LENGTH","MONTH_NAME",
    ]
    drop_replaced_targets = [
            "INJURIES_DIRECT", "INJURIES_INDIRECT",
            "DEATHS_DIRECT", "DEATHS_INDIRECT",
            "DAMAGE_PROPERTY", "DAMAGE_CROPS",
            "INJURIES", "DEATHS",
        ]
    out = out.drop(columns=[column for column in drop_replaced_targets if column in out.columns])
    out = out.drop(columns=[column for column in drop_replaced_features if column in out.columns])
    out = out.loc[~out["EVENT_GROUP"].isin(EXCLUDE_GROUPS)].copy()

    if "CZ_FIPS" in out.columns:
        out["CZ_FIPS"] = out["CZ_FIPS"].astype("string[pyarrow]")

    return out.sort_index(kind="stable")


base_events = build_base_event_frame(df)
print(base_events.shape)
print(base_events.index.min(), base_events.index.max())
base_events.head()
```

```python
########################################################
# Context features: what happened earlier the same day
#######################################################

# A flood on a day with storms all over the country is not the same as an isolated one.
# For every event we summarise the events that **began strictly earlier on the same
# calendar day**:

# `CTX_N_PREV_DAY`: how many events had already started that day (nationwide);
# `CTX_N_GROUP_PREV_DAY`: how many of them belonged to the same event group;
# `CTX_N_STATE_PREV_DAY`: how many of them happened in the same state;
# `CTX_N_HIGH_RISK_PREV_DAY`: how many of them were rated high risk;
# `CTX_N_WIDE_PREV_DAY`: how many of them were regional or widespread in scope;
# `CTX_HOURS_SINCE_STATE_GROUP`: hours since the previous event of the same group
#     in the same state (capped at 48).

# Only the predictors (X) of the earlier events are used, never their casualties (the target, y).
# The features are built one split at a time, so nothing crosses from train to calibration or test.

# ==============================================================================
# CONTEXT FEATURES FROM EARLIER EVENTS OF THE SAME DAY
# ==============================================================================

CONTEXT_COLUMNS = ["CTX_N_PREV_DAY", "CTX_N_GROUP_PREV_DAY", "CTX_N_STATE_PREV_DAY",
                   "CTX_N_HIGH_RISK_PREV_DAY", "CTX_N_WIDE_PREV_DAY",
                   "CTX_HOURS_SINCE_STATE_GROUP"]
CONTEXT_MAX_HOURS = 48.0
WIDE_SCOPES = ["regional", "widespread"]


def add_context_features(frame):
    """Context from events that began strictly EARLIER on the same calendar day.

    Only predictors (X) of the earlier events are used, never their casualties.
    Called on one split at a time, so no information crosses split boundaries."""
    out = frame.copy()
    begin = pd.Series(out.index, index=out.index)

    keys = pd.DataFrame({"day": begin.dt.normalize().to_numpy(),
                         "time": begin.to_numpy(),
                         "group": out["EVENT_GROUP"].to_numpy(),
                         "state": out["STATE"].to_numpy()}, index=out.index)

    high_risk = out["risk"].eq("high").astype("float64")
    wide = out["event_scope"].isin(WIDE_SCOPES).astype("float64")
    one = pd.Series(1.0, index=out.index)

    def running_count(flag, within=()):
        # earlier rows of the same day (and of the same `within` keys), then
        # events with the same start time keep the value of the first of them,
        # so simultaneous events do not count each other
        by = [keys["day"], *(keys[k] for k in within)]
        so_far = flag.groupby(by).cumsum() - flag
        return so_far.groupby([*by, keys["time"]]).transform("min")

    out["CTX_N_PREV_DAY"] = running_count(one)
    out["CTX_N_GROUP_PREV_DAY"] = running_count(one, within=["group"])
    out["CTX_N_STATE_PREV_DAY"] = running_count(one, within=["state"])
    out["CTX_N_HIGH_RISK_PREV_DAY"] = running_count(high_risk)
    out["CTX_N_WIDE_PREV_DAY"] = running_count(wide)

    previous_same_state_group = begin.groupby([keys["state"], keys["group"]]).shift()
    hours = (begin - previous_same_state_group).dt.total_seconds() / 3600.0
    out["CTX_HOURS_SINCE_STATE_GROUP"] = (
        hours.fillna(CONTEXT_MAX_HOURS).clip(lower=0, upper=CONTEXT_MAX_HOURS))

    return out


# one split at a time
train_raw = add_context_features(train_raw)
calibration_raw = add_context_features(calibration_raw)
test_raw = add_context_features(test_raw)

```

```python
# ==============================================================================
# ZERO-VARIANCE FEATURE REMOVAL
# ==============================================================================

categorical_columns = [
    column
    for column in train_raw.columns
    if train_raw[column].dtype in ("string[pyarrow]", "object", "category")
]

numeric_columns = [
    column
    for column in train_raw.columns
    if column not in categorical_columns
]

categorical_zero_variance = [
    column
    for column in categorical_columns
    if train_raw[column].nunique(dropna=False) <= 1]

protected_numeric = set(PREDICTION_TARGETS) | {"EXPOSURE"}
numeric_zero_variance = [
    column
    for column in numeric_columns
    if column not in protected_numeric
    and train_raw[column].nunique(dropna=False) <= 1]

cat=train_raw[categorical_columns]
num=train_raw[numeric_columns]
ordinal_cols = [c for c in cat if c in ORDINAL_LEVELS]        # ranked labels
nominal_cols = [c for c in cat if c not in ORDINAL_LEVELS]    # unranked labels
num_feat = [c for c in num if c not in PREDICTION_TARGETS and c != "EXPOSURE"]


print("=" * 60)
print("Categorical zero-variance variables:", categorical_zero_variance)
print("Numerical zero-variance variables:", numeric_zero_variance)
print("=" * 60)
```

```python
# ==============================================================================
# CRAMER'S V  TEST FOR CATEGORICAL VARIABLES
# ==============================================================================

CRAMER_THRESHOLD = 0.75
PROTECTED = {"WFO","STATE","SOURCE","EVENT_GROUP"}

test_columns = [c for c in categorical_columns if c not in PROTECTED]

rows = []
for a, b in combinations(test_columns, 2):
    tab = pd.crosstab(cat[a], cat[b])
    chi2, p, dof, _ = chi2_contingency(tab)
    k = min(tab.shape) - 1
    v = np.sqrt(chi2 / (tab.to_numpy().sum() * k)) if k else 0.0
    rows.append((a, b, chi2, p, dof, v))

cramers_v = pd.DataFrame(rows, columns=["Variable_1", "Variable_2", "Chi2_Stat", "p_value", "dof", "Cramer_V"]).sort_values("Cramer_V", ascending=False, ignore_index=True)

strong = cramers_v[cramers_v.Cramer_V >= CRAMER_THRESHOLD]
cramer_to_drop = sorted({max((r.Variable_1, r.Variable_2), key=lambda c: cat[c].nunique()) for r in strong.itertuples()})
print("=" * 60, "\nCategorical variables to drop:", cramer_to_drop, "\n" + "=" * 60)
```

```python
# ==============================================================================
# SPEARMAN CORRELATION FILTER FOR NUMERICAL PREDICTORS
# ==============================================================================

SPEARMAN_THRESHOLD = 0.80
PROTECTED_NUMERIC = set(PREDICTION_TARGETS) | {"EXPOSURE"}

correlation_columns = [c for c in numeric_columns if c not in PROTECTED_NUMERIC]

corr = train_raw[correlation_columns].corr(method="spearman").abs()
mask = np.triu(np.ones(corr.shape, dtype=bool), k=1)

corr_to_drop = corr.columns[(corr.where(mask) > SPEARMAN_THRESHOLD).any()].tolist()

print("=" * 60, "\nNumerical predictor variables to drop:", corr_to_drop, "\n" + "=" * 60)
```

```python
# ==============================================================================
# KENDALL'S TAU-b TEST — NUMERICAL vs ORDINAL CATEGORICAL VARIABLES
# ==============================================================================

KENDALL_THRESHOLD = 0.75

ordinal_num = pd.DataFrame(index=train_raw.index)

for col in ordinal_cols:
    levels = ORDINAL_LEVELS[col]
    encoding = {}
    for position, level in enumerate(levels):
        encoding[level] = position
    ordinal_num[col] = train_raw[col].map(encoding)

X = train_raw[num_feat].to_numpy(dtype=float)
Y = ordinal_num.to_numpy(dtype=float)

n_num, n_ord = X.shape[1], Y.shape[1]
tau_mat = np.empty((n_num, n_ord))
pval_mat = np.empty((n_num, n_ord))
count_mat = np.empty((n_num, n_ord), dtype=int)

valid_x = ~np.isnan(X)
valid_y = ~np.isnan(Y)

for i in range(n_num):
    for j in range(n_ord):
        ok = valid_x[:, i] & valid_y[:, j]
        count_mat[i, j] = ok.sum()
        tau_mat[i, j], pval_mat[i, j] = kendalltau(X[ok, i], Y[ok, j], variant="b")

df_tau = pd.DataFrame({
    "Numerical":   np.repeat(list(num_feat), n_ord),
    "Categorical": np.tile(list(ordinal_cols), n_num),
    "tau_b":       tau_mat.ravel(),
    "p_value":     pval_mat.ravel(),
    "n_used":      count_mat.ravel()})
df_tau["Assoc"] = df_tau["tau_b"].abs()
df_tau = df_tau.sort_values("Assoc", ascending=False, ignore_index=True)

print("Strongest rank agreements (|tau_b| near 1 = duplicated information):")
print(df_tau.head(10).to_string(index=False))

strong = df_tau[df_tau["Assoc"] >= KENDALL_THRESHOLD]

df_tau_to_drop = sorted(set(strong["Categorical"]))
```

```python
# ==============================================================================
# KRUSKAL-WALLIS H-TEST — NUMERICAL vs NOMINAL CATEGORICAL VARIABLES
# ==============================================================================

KRUSKAL_ETA_THRESHOLD = 0.75
PROTECTED = {"WFO","STATE","SOURCE","EVENT_GROUP"}

nominal_cols = [c for c in nominal_cols if c not in PROTECTED]

cat_codes = {}
for col in nominal_cols:
    cat_codes[col] = pd.factorize(train_raw[col])[0]

X = train_raw[num_feat].to_numpy(dtype=float)
valid_x = ~np.isnan(X)

def tie_correction(ranks):
    n = ranks.size
    if n < 2:
        return 0.0
    counts = np.unique(ranks, return_counts=True)[1].astype(float)
    tied = counts[counts > 1]
    return 1.0 - (tied ** 3 - tied).sum() / (n ** 3 - n)


def kruskal_from_ranks(ranks, group_codes, tie_corr):
    n = ranks.size
    group_size = np.bincount(group_codes)
    rank_sum = np.bincount(group_codes, weights=ranks)

    non_empty = group_size > 0
    group_size = group_size[non_empty]
    rank_sum = rank_sum[non_empty]
    k = group_size.size

    if k < 2 or n <= k or tie_corr <= 0:
        return np.nan, np.nan, k

    H = (12.0 / (n * (n + 1)) * np.sum(rank_sum ** 2 / group_size) - 3.0 * (n + 1)) / tie_corr
    return H, chi2_dist.sf(H, k - 1), k

rows = []

for i, num_col in enumerate(num_feat):
    x = X[:, i]
    ok_x = valid_x[:, i]
    n_x = int(ok_x.sum())

    ranks_cached = None
    tie_cached = None

    for cat_col in nominal_cols:
        codes = cat_codes[cat_col]
        ok = ok_x & (codes >= 0)
        n = int(ok.sum())
        if n < 3:
            continue

        if n == n_x:
            if ranks_cached is None:
                ranks_cached = rankdata(x[ok])
                tie_cached = tie_correction(ranks_cached)
            ranks, tie_corr = ranks_cached, tie_cached
        else:
            ranks = rankdata(x[ok])
            tie_corr = tie_correction(ranks)

        H, p, k = kruskal_from_ranks(ranks, codes[ok], tie_corr)
        if np.isnan(H):
            continue

        eta_sq = max(0.0, (H - k + 1) / (n - k))
        rows.append((num_col, cat_col, H, p, n, k, eta_sq))

df_kw = pd.DataFrame(rows,columns=["Numerical", "Categorical", "H_Stat", "p_value", "n_used", "k_groups",
                                   "Eta_Sq"]).sort_values("Eta_Sq", ascending=False, ignore_index=True)

print("Strongest Kruskal-Wallis associations (ranked by Eta-Squared):")
print(df_kw.head(10).to_string(index=False))

strong = df_kw[df_kw["Eta_Sq"] >= KRUSKAL_ETA_THRESHOLD]
df_kw_to_drop = sorted(set(strong["Categorical"]))
```

```python
#####################################
# Encode to tensors
#####################################

def fit_encoders(train_frame):
    # keep numeric columns
    numeric_columns = SCALAR + EPISODE_EMB + EVENT_EMB
    training_numeric_values = train_frame[numeric_columns].to_numpy(
        dtype="float64",
        copy=True
    )

    # log only on log_scalar
    for column_position, column_name in enumerate(numeric_columns):
        if column_name in LOG_SCALAR:
            original_values = training_numeric_values[:, column_position]
            non_negative_values = original_values.clip(min=0.0)
            logarithmic_values = np.log1p(non_negative_values)
            training_numeric_values[:, column_position] = logarithmic_values

    # Statistics for standardization on numeric
    numeric_mean = training_numeric_values.mean(axis=0)
    numeric_std = training_numeric_values.std(axis=0, ddof=1)

    # Statistics for standardization on seasonal
    training_seasonal_values = train_frame[SEASONAL_COLUMNS].to_numpy(
        dtype="float64"
    )

    seasonal_mean = training_seasonal_values.mean(axis=0)
    seasonal_std = training_seasonal_values.std(axis=0, ddof=1)

    # Statistics for standardization on YEAR
    year_mean = float(train_frame[EPOCH_COLUMN].mean())
    year_std = float(train_frame[EPOCH_COLUMN].std())

    # dictionary for categorical
    vocabularies = {}

    for column_name in CATEGORICAL:
        observed_categories = train_frame[column_name].astype("object").unique()
        sorted_categories = sorted(observed_categories)
        category_codes = {}

        for category_position, category_name in enumerate(sorted_categories):
            # zero for unseen categories
            category_codes[category_name] = category_position + 1
        vocabularies[column_name] = category_codes

    # encoders dictionary
    encoders = {}
    encoders["vocabularies"] = vocabularies
    encoders["numeric_columns"] = numeric_columns
    encoders["mean"] = numeric_mean
    encoders["std"] = numeric_std
    encoders["seasonal_mean"] = seasonal_mean
    encoders["seasonal_std"] = seasonal_std
    encoders["year_mean"] = year_mean
    encoders["year_std"] = year_std

    return encoders

def encode_split(frame, encoders):
    # keep values from encoders
    vocabularies = encoders["vocabularies"]
    numeric_columns = encoders["numeric_columns"]
    numeric_mean = encoders["mean"]
    numeric_std = encoders["std"]
    seasonal_mean = encoders["seasonal_mean"]
    seasonal_std = encoders["seasonal_std"]
    year_mean = encoders["year_mean"]
    year_std = encoders["year_std"]

    # --------------------------------------------------
    # Categorical variables preparation.
    # --------------------------------------------------

    categorical_arrays = []
    for column_name in CATEGORICAL:

        original_categories = frame[column_name].astype("object")
        column_vocabulary = vocabularies[column_name]
        encoded_categories = original_categories.map(column_vocabulary)
        encoded_categories = encoded_categories.fillna(0)
        category_array = encoded_categories.to_numpy(dtype="int64")
        categorical_arrays.append(category_array)

    categorical_values = np.column_stack(categorical_arrays)

    # --------------------------------------------------
    # Numerical variables preparation.
    # --------------------------------------------------

    numeric_values = frame[numeric_columns].to_numpy(
        dtype="float64",
        copy=True
    )


    for column_position, column_name in enumerate(numeric_columns):
        if column_name in LOG_SCALAR:
            original_values = numeric_values[:, column_position]
            non_negative_values = original_values.clip(min=0.0)
            logarithmic_values = np.log1p(non_negative_values)
            numeric_values[:, column_position] = logarithmic_values

    # Standardization.
    centered_numeric_values = numeric_values - numeric_mean
    standardized_numeric_values = centered_numeric_values / numeric_std

    # --------------------------------------------------
    # Seasonal preparation.
    # --------------------------------------------------

    seasonal_values = frame[SEASONAL_COLUMNS].to_numpy(dtype="float64")
    centered_seasonal_values = seasonal_values - seasonal_mean
    standardized_seasonal_values = centered_seasonal_values / seasonal_std

    # --------------------------------------------------
    # YEAR preparation.
    # --------------------------------------------------

    year_values = frame[EPOCH_COLUMN].to_numpy(dtype="float64")
    centered_year_values = year_values - year_mean
    standardized_year_values = centered_year_values / year_std
    standardized_year_values = standardized_year_values.reshape(-1, 1)

    # --------------------------------------------------
    # Exposure and target.
    # --------------------------------------------------

    exposure_values = frame[[EXPOSURE_COL]].to_numpy(dtype="float32")
    target_values = frame[PREDICTION_TARGETS].to_numpy(dtype="float32")

    # --------------------------------------------------
    # Transformation in tensors.
    # --------------------------------------------------

    encoded_data = {}
    encoded_data["categorical"] = torch.tensor(
        categorical_values,
        dtype=torch.long,
        device=DEVICE
    )
    encoded_data["numeric"] = torch.tensor(
        standardized_numeric_values,
        dtype=torch.float32,
        device=DEVICE
    )
    encoded_data["seasonal"] = torch.tensor(
        standardized_seasonal_values,
        dtype=torch.float32,
        device=DEVICE
    )
    encoded_data["epoch"] = torch.tensor(
        standardized_year_values,
        dtype=torch.float32,
        device=DEVICE
    )
    encoded_data["exposure"] = torch.tensor(
        exposure_values,
        dtype=torch.float32,
        device=DEVICE
    )
    encoded_data["y"] = torch.tensor(
        target_values,
        dtype=torch.float32,
        device=DEVICE
    )
    return encoded_data


training_data = splits_raw["train"]

def encoded_cardinalities(encoders):
    """Rows of each categorical embedding table: known categories plus the unseen slot."""
    return [len(encoders["vocabularies"][column_name]) + 1 for column_name in CATEGORICAL]


# encoders of the final model, fitted on its training years (1950-2023) alone
ENCODERS = fit_encoders(train_frame=splits_raw["train"])

def encoded_cardinalities(encoders):
    """Number of embedding rows for each categorical variable."""
    return [
        len(encoders["vocabularies"][column_name]) + 1
        for column_name in CATEGORICAL
    ]


def fit_and_encode_splits(raw_splits):
    """
    Fit encoders on the training split only,
    then apply them to every split.
    """

    # Fit preprocessing only on training data
    encoders = fit_encoders(
        train_frame=raw_splits["train"]
    )

    # Apply the fitted preprocessing to all splits
    encoded_splits = {}

    for split_name, split_frame in raw_splits.items():
        encoded_splits[split_name] = encode_split(
            frame=split_frame,
            encoders=encoders
        )

    # Embedding-table sizes for categorical variables
    cardinalities = encoded_cardinalities(encoders)

    return encoders, encoded_splits, cardinalities


# --------------------------------------------------
# Final model
# Encoders fitted on 1950-2023
# --------------------------------------------------

ENCODERS, SPLITS, CARDINALITIES = fit_and_encode_splits(
    raw_splits=splits_raw
)


# --------------------------------------------------
# Early-stopping model
# Encoders fitted on 1950-2022
# 2023 remains unseen
# --------------------------------------------------

(
    EARLY_STOPPING_ENCODERS,
    EARLY_STOPPING_SPLITS,
    EARLY_STOPPING_CARDINALITIES,
) = fit_and_encode_splits(
    raw_splits=early_stopping_raw
)


gc.collect()


# --------------------------------------------------
# Diagnostics
# --------------------------------------------------

levels_per_category = dict(
    zip(CATEGORICAL, CARDINALITIES)
)

print("levels per categorical:", levels_per_category)


block_names = (
    "categorical",
    "numeric",
    "seasonal",
    "epoch",
    "exposure",
)

block_widths = {}

for block_name in block_names:
    training_tensor = SPLITS["train"][block_name]
    block_widths[block_name] = tuple(training_tensor.shape[1:])

print("block widths:", block_widths)


```

```python
####################################
# Feature tokens
####################################


MODULES = []   # everything from transformation, embedding, attentions....
TENSORS = []   # learnable tensors parameters


def make_linear(in_features, out_features):
    layer = nn.Linear(in_features, out_features).to(DEVICE)
    nn.init.xavier_uniform_(layer.weight)
    nn.init.zeros_(layer.bias)
    MODULES.append(layer)
    return layer


def make_embedding(n_levels, width):
    table = nn.Embedding(n_levels, width).to(DEVICE)
    nn.init.normal_(table.weight, std=0.1)
    MODULES.append(table)
    return table

def periodic_embedding(values, frequencies):
    """[sin(2 pi c x), cos(2 pi c x)] for a bank of learned frequencies c."""
    angles = 2.0 * np.pi * values.unsqueeze(-1) * frequencies
    return torch.cat([torch.sin(angles), torch.cos(angles)], dim=-1)


def make_norm(width):
    layer = nn.LayerNorm(width).to(DEVICE)
    MODULES.append(layer)
    return layer


def make_attention(config):
    layer = nn.MultiheadAttention(config["d_model"], config["n_heads"],
                                  dropout=config["dropout"], batch_first=True).to(DEVICE)
    MODULES.append(layer)
    return layer


def make_tensor(values):
    tensor = nn.Parameter(values.to(DEVICE))
    TENSORS.append(tensor)
    return tensor


def make_encoder_block(config):
    d = config["d_model"]
    return {
        "norm_attention": make_norm(d),
        "attention": make_attention(config),
        "norm_ffn": make_norm(d),
        "ffn_in": make_linear(d, d * config["ffn_multiplier"]),
        "ffn_out": make_linear(d * config["ffn_multiplier"], d),
    }


def make_params(config, cardinalities, baseline_rate):
    """Every trainable object of one target's model, in one dictionary."""
    d = config["d_model"]
    MODULES.clear()
    TENSORS.clear()
    torch.manual_seed(config["seed"])

    categorical_tables, categorical_projections = {}, {}
    for position, name in enumerate(CATEGORICAL):
        categorical_tables[name] = make_embedding(cardinalities[position],
                                                  2 if name in RANK_TWO else d)
        categorical_projections[name] = make_linear(2, d) if name in RANK_TWO else nn.Identity()

    params = {
        "config": config,
        "categorical_tables": categorical_tables,
        "categorical_projections": categorical_projections,
        "scalar_frequencies": {name: make_tensor(torch.randn(config["n_frequencies"]) * 0.5)
                               for name in SCALAR},
        "scalar_projections": {name: make_linear(2 * config["n_frequencies"], d)
                               for name in SCALAR},
        "episode_projection": make_linear(len(EPISODE_EMB), d),
        "event_projection": make_linear(len(EVENT_EMB), d),
        "seasonal_projection": make_linear(len(SEASONAL_COLUMNS), d),
        "epoch_projection": make_linear(1, d),
        "cls_token": make_tensor(torch.randn(1, 1, d) * 0.02),
        "column_identity": make_tensor(torch.randn(1, len(TOKEN_NAMES), d) * 0.02),
        "encoder_blocks": [make_encoder_block(config) for _ in range(config["n_layers"])],
        "encoder_norm_out": make_norm(d),
        "decoder_attention": make_attention(config),
        "decoder_norm_ffn": make_norm(d),
        "decoder_ffn_in": make_linear(d, d * config["ffn_multiplier"]),
        "decoder_ffn_out": make_linear(d * config["ffn_multiplier"], d),
        "decoder_norm_out": make_norm(d),
        "head_weight": make_tensor(torch.zeros(1, d)),
        "head_bias": make_tensor(torch.log(baseline_rate)),
    }
    params["modules"] = list(MODULES)
    params["tensors"] = list(TENSORS)
    return params


def all_parameters(params):
    """Flat list of every trainable tensor of this model, in a fixed order."""
    tensors = list(params["tensors"])
    for module in params["modules"]:
        tensors.extend(module.parameters())
    return tensors


def set_mode(params, training):
    """Switch dropout on or off, since there is no nn.Module wrapper to do it."""
    for module in params["modules"]:
        module.train(training)


def build_tokens(params, batch):
    """One token per predictor, in the order of TOKEN_NAMES. Shape (rows, tokens, d_model)."""
    tokens = [params["cls_token"].expand(batch["categorical"].shape[0], -1, -1)]

    for position, name in enumerate(CATEGORICAL):
        embedded = params["categorical_tables"][name](batch["categorical"][:, position])
        tokens.append(params["categorical_projections"][name](embedded).unsqueeze(1))

    for position, name in enumerate(SCALAR):
        features = periodic_embedding(batch["numeric"][:, position],
                                      params["scalar_frequencies"][name])
        tokens.append(params["scalar_projections"][name](features).unsqueeze(1))

    start = len(SCALAR)
    tokens.append(params["episode_projection"](
        batch["numeric"][:, start:start + len(EPISODE_EMB)]).unsqueeze(1))
    start += len(EPISODE_EMB)
    tokens.append(params["event_projection"](
        batch["numeric"][:, start:start + len(EVENT_EMB)]).unsqueeze(1))

    tokens.append(params["seasonal_projection"](batch["seasonal"]).unsqueeze(1))
    tokens.append(params["epoch_projection"](batch["epoch"]).unsqueeze(1))

    return torch.cat(tokens, dim=1) + params["column_identity"]


```

```python
###################################
# Encoder, Decoder, Poisson head
###################################

def run_encoder(params, tokens, keep_attention=False):
    """run tokens by encoder blocks."""

    attention_maps = []

    for block in params["encoder_blocks"]:

        # --------------------------------------------------
        # SELF-ATTENTION
        # --------------------------------------------------

        # Normalization before attention.
        normalized_tokens = block["norm_attention"](tokens)

        # Query, key and value from tokens.
        attention_output, attention_weights = block["attention"](
            query=normalized_tokens,
            key=normalized_tokens,
            value=normalized_tokens,
            need_weights=keep_attention,
            average_attn_weights=True
        )

        # update tokens with attention
        tokens = tokens + attention_output

        # attention maps.
        if keep_attention:
            attention_maps.append(attention_weights)

        # --------------------------------------------------
        # FEED-FORWARD NN
        # --------------------------------------------------


        normalized_tokens = block["norm_ffn"](tokens)
        expanded_values = block["ffn_in"](normalized_tokens)
        activated_values = F.gelu(expanded_values)
        ffn_output = block["ffn_out"](activated_values)
        tokens = tokens + ffn_output

    # Final normalization after the entire pre-norm encoder stack.
    # LayerNorm acts separately on the d_model components of each token.
    memory = params["encoder_norm_out"](tokens)
    return memory, attention_maps


def run_decoder(params, memory, keep_attention=False):
    """Use the encoded CLS to read the other encoded tokens (no learned query)."""

    # Preserve the token axis: (batch, 1, d_model).
    # This CLS is contextualized by the encoder and varies by event.
    queries = memory[:, :1, :]
    feature_memory = memory[:, 1:, :]

    # --------------------------------------------------
    # CROSS-ATTENTION
    # --------------------------------------------------

    # CLS and feature memory already passed through encoder_norm_out.
    # Project Q directly from CLS, without a second query normalization.
    attention_output, attention_weights = params["decoder_attention"](
        query=queries,
        key=feature_memory,
        value=feature_memory,
        need_weights=keep_attention,
        average_attn_weights=True
    )

    # update query with attention
    queries = queries + attention_output

    # --------------------------------------------------
    # FEED-FORWARD NN
    # --------------------------------------------------

    # Normalize the updated CLS before the feed-forward network.
    normalized_queries = params["decoder_norm_ffn"](queries)
    expanded_values = params["decoder_ffn_in"](normalized_queries)
    activated_values = F.gelu(expanded_values)
    ffn_output = params["decoder_ffn_out"](activated_values)

    # output query decoder.
    queries = queries + ffn_output

    # Final normalization after the decoder residuals, before the Poisson head.
    queries = params["decoder_norm_out"](queries)
    return queries, attention_weights


def apply_head(params, queries, exposure):
    """decoder output transformed in log expected count (exp is applied in predict)."""


    head_weights = params["head_weight"]
    head_bias = params["head_bias"]
    head_weights = head_weights.unsqueeze(0)
    weighted_components = queries * head_weights
    weighted_sum = weighted_components.sum(dim=-1)
    log_rate = weighted_sum + head_bias
    log_exposure = torch.log(exposure)
    log_expected_count = log_rate + log_exposure

    return log_expected_count


def forward_from_tokens(params, tokens, exposure, keep_attention=False):
    """encoder, decoder and head starting from built tokens."""

    # token fed into encoder.
    memory, encoder_attention = run_encoder(
        params=params,
        tokens=tokens,
        keep_attention=keep_attention
    )

    # decoder read memory.
    queries, decoder_attention = run_decoder(
        params=params,
        memory=memory,
        keep_attention=keep_attention
    )

    # head make prediction.
    predictions = apply_head(
        params=params,
        queries=queries,
        exposure=exposure
    )

    # collection attention maps.
    attention_information = {}
    attention_information["encoder"] = encoder_attention
    attention_information["decoder"] = decoder_attention

    return predictions, attention_information


def forward(params, batch, keep_attention=False):
    """end-to-end pipeline"""

    # build tokens
    tokens = build_tokens(
        params=params,
        batch=batch
    )
    exposure = batch["exposure"]

    # run the model.
    predictions, attention_information = forward_from_tokens(
        params=params,
        tokens=tokens,
        exposure=exposure,
        keep_attention=keep_attention
    )

    return predictions, attention_information
```

### Training

```python
#############################################################
# FINE TUNING BY EXPANDING-YEAR CROSS VALIDATION
#############################################################

BACKTEST_VALIDATION_YEARS = (2019, 2020, 2021, 2022, 2023)

# network and optimiser
MODEL_CONFIG = {
    "d_model": 64,
    "n_heads": 4,
    "n_layers": 3,
    "ffn_multiplier": 2,
    "dropout": 0.1,
    "n_frequencies": 8,
   "learning_rate": 2e-3,
    "weight_decay": 1e-4,
    "batch_size": 512,
    "max_epochs": 60,
    "patience": 5,
    "seed": DEFAULT_RANDOM_STATE,
}

PREDICT_BATCH_SIZE = 16384


TUNING_CONFIG = {
    "n_trials": 15,
    "max_epochs": 40,
    "patience": 5,
}


def expanding_year_splits(frame, validation_years=BACKTEST_VALIDATION_YEARS):
    """Train on every year strictly before val_year, validate on val_year alone."""
    years = frame.index.year.to_numpy()
    for fold, val_year in enumerate(validation_years, start=1):
        train_index = np.flatnonzero(years < val_year)
        val_index = np.flatnonzero(years == val_year)
        if len(train_index) == 0 or len(val_index) == 0:
            continue
        yield fold, val_year, train_index, val_index


def prepare_fold_views(frame, target, validation_years=BACKTEST_VALIDATION_YEARS):
    """Encoders fitted on each fold's train years only; tensors built once and reused by every trial."""
    index = PREDICTION_TARGETS.index(target)
    folds = []
    for fold, val_year, train_index, val_index in expanding_year_splits(frame, validation_years):
        fold_train = frame.iloc[train_index]
        encoders = fit_encoders(fold_train)
        tensors = {"train": encode_split(fold_train, encoders),
                   "validation": encode_split(frame.iloc[val_index], encoders)}
        views = {name: target_view(part, index) for name, part in tensors.items()}
        folds.append({
            "fold": fold,
            "validation_year": val_year,
            "views": views,
            "cardinalities": [len(encoders["vocabularies"][column]) + 1 for column in CATEGORICAL],
            "baseline_rate": (views["train"]["y"].sum(dim=0)
                              / views["train"]["exposure"].sum()).clamp(min=1e-8),
        })
        print(f"fold {fold}: train <= {val_year - 1} ({len(train_index):,} rows)   "
              f"validation = {val_year} ({len(val_index):,} rows)")
    return folds


def suggest_config(trial, base_config=MODEL_CONFIG, tuning=TUNING_CONFIG):
    """Search space over network and optimiser; seed is kept from the base config."""
    config = dict(base_config)
    config.update({
        "d_model": trial.suggest_categorical("d_model", [64, 128, 256]),
        "n_heads": trial.suggest_categorical("n_heads", [4, 8]),   # both divide every d_model
        "n_layers": trial.suggest_int("n_layers", 1, 4),
        "ffn_multiplier": trial.suggest_categorical("ffn_multiplier", [2, 4]),
        "dropout": trial.suggest_float("dropout", 0.0, 0.5),
        "n_frequencies": trial.suggest_categorical("n_frequencies", [8, 16, 32, 64]),
        "learning_rate": trial.suggest_float("learning_rate", 1e-4, 3e-3, log=True),
        "weight_decay": trial.suggest_float("weight_decay", 1e-6, 1e-3, log=True),
        "batch_size": trial.suggest_categorical("batch_size", [256, 512, 1024]),
        "max_epochs": tuning["max_epochs"],
        "patience": tuning["patience"],
    })
    return config


def fold_score(config, fold):
    """Train on one fold with early stopping on its validation year; return best MPD and best epoch."""
    params = make_params(config, fold["cardinalities"], fold["baseline_rate"])
    try:
        with contextlib.redirect_stdout(io.StringIO()):   # silence the per-epoch prints
            history, best_epoch = train_model(params, fold["views"]["train"], config,
                                              epochs=config["max_epochs"],
                                              validation_split=fold["views"]["validation"],
                                              patience=config["patience"])
        best_deviance = float(history["val_MPD"].min())
        if not np.isfinite(best_deviance):
            raise ValueError("validation deviance is not finite")
        return best_deviance, best_epoch
    finally:
        del params
        gc.collect()
        torch.cuda.empty_cache()


def tune_config_cv(frame, target, validation_years=BACKTEST_VALIDATION_YEARS, tuning=TUNING_CONFIG):
    """Optuna search over expanding-window folds, minimizing mean Poisson deviance.
    Returns MODEL_CONFIG updated with the best parameters, and the study."""
    optuna.logging.set_verbosity(optuna.logging.WARNING)

    print(f"Precomputing folds for {target}...")
    folds = prepare_fold_views(frame, target, validation_years)

    def objective(trial):
        config = suggest_config(trial, tuning=tuning)
        scores, epochs = [], []
        started = time.time()
        try:
            for step, fold in enumerate(folds):
                score, best_epoch = fold_score(config, fold)
                scores.append(score)
                epochs.append(best_epoch)
                print(f"    trial {trial.number} fold {step + 1}/{len(folds)}  MPD {score:.5f}  "
                      f"best epoch {best_epoch}  ({(time.time() - started) / 60:.1f} min)", flush=True)
                trial.report(float(np.mean(scores)), step)
                if trial.should_prune():
                    raise optuna.TrialPruned()
        except (ValueError, torch.cuda.OutOfMemoryError) as error:
            gc.collect()
            torch.cuda.empty_cache()
            raise optuna.TrialPruned(f"numerical instability or out of memory: {error}")

        trial.set_user_attr("fold_MPD", scores)
        trial.set_user_attr("best_epochs", epochs)
        print(f"  trial {trial.number:3d}  mean MPD {np.mean(scores):.5f}  "
              f"best epochs {epochs}  ({(time.time() - started) / 60:.1f} min)")
        return float(np.mean(scores))

    study = optuna.create_study(
        direction="minimize",
        sampler=optuna.samplers.TPESampler(seed=DEFAULT_RANDOM_STATE),
        pruner=optuna.pruners.MedianPruner(n_startup_trials=5, n_warmup_steps=1),
    )
    study.optimize(objective, n_trials=tuning["n_trials"], n_jobs=1)

    best_config = dict(MODEL_CONFIG)
    best_config.update(study.best_params)

    del folds
    gc.collect()
    torch.cuda.empty_cache()
    return best_config, study


def find_best_config_for_all_targets(frame, targets, validation_years=BACKTEST_VALIDATION_YEARS,
                                     tuning=TUNING_CONFIG):
    """One tuned configuration per target."""
    best_configs, studies = {}, {}
    for target in targets:
        print(f"Tuning tabular transformer for target {target}...")
        best_configs[target], studies[target] = tune_config_cv(frame, target, validation_years, tuning)
        print(f"  best configuration: {best_configs[target]}")
    return best_configs, studies
```

```python
poisson_loss=nn.PoissonNLLLoss(
    log_input=True,
    full=False,
    reduction="mean",
    eps=1e-7
)


def select_rows_from_split(complete_split, row_positions):
    selected_batch = {}
    for tensor_name, complete_tensor in complete_split.items():
        selected_tensor = complete_tensor[row_positions]
        selected_batch[tensor_name] = selected_tensor
    return selected_batch


def target_view(split, index):
    view = dict(split)
    view["y"] = split["y"][:, index:index + 1]
    return view


def predict(params, split, batch_size=PREDICT_BATCH_SIZE):
    set_mode(params, False)
    rates = []
    n_rows = len(split["y"])
    with torch.no_grad():
        for start in range(0, n_rows, batch_size):
            batch = select_rows_from_split(split, slice(start, min(start + batch_size, n_rows)))
            log_rate, _ = forward(params, batch,keep_attention=False)
            rates.append(torch.exp(log_rate).cpu())
    return torch.cat(rates).numpy()


def compute_metrics(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    return {
        "n": len(y_true),
        "events": int((y_true > 0).sum()),
        "D2": np.nan if y_true.sum() == 0 else d2_tweedie_score(y_true, y_pred, power=1),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "MPD": mean_poisson_deviance(y_true, y_pred),
    }


def snapshot_parameters(params):
    return [tensor.detach().clone() for tensor in all_parameters(params)]


def restore_parameters(params, snapshot):
    with torch.no_grad():
        for tensor, saved in zip(all_parameters(params), snapshot):
            tensor.copy_(saved)


def train_model(params, train_split, config, epochs, validation_split=None, patience=None):
    """One target's training loop."""
    trainable = all_parameters(params)
    optimizer = torch.optim.AdamW(trainable, lr=config["learning_rate"],
                                  weight_decay=config["weight_decay"])
    n_rows = len(train_split["y"])
    steps_per_epoch = math.ceil(n_rows / config["batch_size"])
    schedule_epochs = max(epochs, config["max_epochs"])
    schedule = torch.optim.lr_scheduler.CosineAnnealingLR(
        optimizer, T_max=schedule_epochs * steps_per_epoch)
    generator = torch.Generator(device="cpu").manual_seed(config["seed"])

    observed = None if validation_split is None else validation_split["y"][:, 0].cpu().numpy()
    history, best_deviance, best_epoch, best_state, waited = [], np.inf, epochs, None, 0

    for epoch in range(epochs):
        set_mode(params, True)
        order = torch.randperm(n_rows, generator=generator).to(DEVICE)
        running = torch.zeros((), device=DEVICE)

        for start in range(0, n_rows, config["batch_size"]):
            positions = order[start:start + config["batch_size"]]
            batch = select_rows_from_split(train_split, positions)
            log_rate, _ = forward(params, batch,keep_attention=False)
            loss = poisson_loss(log_rate, batch["y"])

            optimizer.zero_grad(set_to_none=True)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(trainable, 1.0)
            optimizer.step()
            schedule.step()
            running += loss.detach() * positions.numel()

        record = {"epoch": epoch + 1, "loss": float(running) / n_rows}

        if validation_split is None:
            history.append(record)
            print(f"  epoch {epoch + 1:3d}  loss {record['loss']:9.5f}")
            continue

        scores = compute_metrics(observed, predict(params, validation_split)[:, 0])
        record["val_MPD"] = scores["MPD"]
        record["val_D2"] = scores["D2"]
        history.append(record)
        print(f"  epoch {epoch + 1:3d}  loss {record['loss']:9.5f}  "
              f"validation MPD {scores['MPD']:9.5f}  D2 {scores['D2']:7.4f}")

        if scores["MPD"] < best_deviance:
            best_deviance, best_epoch, best_state, waited = (
                scores["MPD"], epoch + 1, snapshot_parameters(params), 0)
        else:
            waited += 1
            if patience is not None and waited >= patience:
                print(f"  stopped at epoch {epoch + 1}; best was epoch {best_epoch}")
                break

    if best_state is not None:
        restore_parameters(params, best_state)

    return pd.DataFrame(history), best_epoch

def baseline_rate_of(view):
    """Average events per unit of exposure on the fitted rows: where the head starts."""
    return (view["y"].sum(dim=0) / view["exposure"].sum()).clamp(min=1e-8)


def choose_number_of_epochs(config, target):
    """First pass: train on 1950-2022 and stop when 2023 stops improving.
    Only the number of epochs is kept from this pass; the model itself is discarded."""
    index = PREDICTION_TARGETS.index(target)
    views = {name: target_view(split, index) for name, split in EARLY_STOPPING_SPLITS.items()}
    params = make_params(config, EARLY_STOPPING_CARDINALITIES, baseline_rate_of(views["train"]))

    history, best_epoch = train_model(params, views["train"], config,
                                      epochs=config["max_epochs"],
                                      validation_split=views["validation"],
                                      patience=config["patience"])
    del params
    gc.collect()
    torch.cuda.empty_cache()
    return history, best_epoch


def run_target_model(splits, config, target):
    """Two passes: 2023 decides the number of epochs, then the model is refitted on
    1950-2023, the same years of other models."""
    print("first pass: choosing the number of epochs on 2023")
    history, best_epoch = choose_number_of_epochs(config, target)

    index = PREDICTION_TARGETS.index(target)
    views = {name: target_view(split, index) for name, split in splits.items()}
    params = make_params(config, CARDINALITIES, baseline_rate_of(views["train"]))

    print(f"second pass: refitting on 1950-2023 for {best_epoch} epochs   "
          f"trainable parameters {sum(int(t.numel()) for t in all_parameters(params)):,}")
    train_model(params, views["train"], config, epochs=best_epoch)

    return {"target": target, "params": params, "views": views,
            "history": history, "best_epoch": best_epoch}

```

```python
###########################################
# Adaptive Conformal Prediction
###########################################


def finite_sample_conformal_quantile(scores, alpha):
    """Computes the finite-sample-corrected (1-alpha)
    quantile of the calibration scores."""
    scores = np.sort(np.asarray(scores, dtype=float))
    rank = int(np.ceil((len(scores) + 1) * (1.0 - alpha)))

    if rank > len(scores):
        return np.inf

    return float(scores[rank - 1])


def adaptive_conformal_by_period(
    y_true,
    y_pred,
    dates,
    calibration_scores,
    alpha_target=0.10,
    gamma=0.01,
):
    """Adaptive conformal intervals updated once per temporal period."""
    actual = np.asarray(y_true, dtype=float)
    predicted = np.asarray(y_pred, dtype=float)
    period_dates = pd.DatetimeIndex(dates).normalize()

    lower = np.empty(len(actual))
    upper = np.empty(len(actual))
    covered = np.empty(len(actual), dtype=bool)
    alphas = np.empty(len(actual))

    alpha_t = alpha_target
    min_alpha = max(
        1e-3,
        1.0 / (len(calibration_scores) + 1),
    )

    for period in np.unique(period_dates):
        positions = np.flatnonzero(period_dates == period)

        alpha_t = float(np.clip(alpha_t, min_alpha, 0.999))
        q = finite_sample_conformal_quantile(calibration_scores, alpha_t)

        margin = q * np.sqrt(np.clip(predicted[positions], 1, None))
        lower[positions] = np.maximum(0.0, predicted[positions] - margin)
        upper[positions] = predicted[positions] + margin
        covered[positions] = (
            (actual[positions] >= lower[positions])
            & (actual[positions] <= upper[positions])
        )
        alphas[positions] = alpha_t

        alpha_t += gamma * (
            alpha_target - (1.0 - covered[positions].mean())
        )

    return lower, upper, upper - lower, covered, alphas


def _aggregate_split_predictions(
    predictions,
    split,
    freq,
    group_column=None,
    group_name=None,
):
    """Filters predictions to the given split and resamples
    the actual/predicted values by summing them at the given time."""
    split_mask = predictions["split"].eq(split)
    split_data = predictions.loc[
        split_mask,
        ["actual", "predicted"],
    ].copy()

    if group_name is not None:
        group_values = predictions.loc[split_mask, group_column]
        group_mask = group_values.eq(group_name).to_numpy()
        split_data.loc[~group_mask, ["actual", "predicted"]] = 0.0

    return split_data.resample(freq).sum()


def build_daily_base(predictions, group_column=None, group_name=None):
    """THE base measure: daily totals of actual and predicted, one row per
    calendar day, for the calibration and test splits."""
    return {
        split: _aggregate_split_predictions(
            predictions,
            split=split,
            freq="D",
            group_column=group_column,
            group_name=group_name,
        )
        for split in ("calibration", "test")
    }


CALIBRATION_WINDOW_DAYS = {"D": 1, "MS": 30}


def conformity_scores(daily_calibration, window_days):
    """Poisson-standardized absolute residuals of the daily base, summed over
    a rolling window as wide as the view will be read at.

    The window is what makes the monthly view usable."""
    frame = daily_calibration
    if window_days > 1:
        frame = frame.rolling(window_days).sum().dropna()

    return np.abs(
        frame["actual"].to_numpy() - frame["predicted"].to_numpy()
    ) / np.sqrt(np.clip(frame["predicted"].to_numpy(), 1, None))


def conformalize_base(
    daily_base,
    freq="D",
    alpha_target=0.10,
    gamma=0.01,
    calibration_window=None,
):
    """Adaptive conformal intervals on the daily base, re-aggregated to a
    monthly view."""
    calibration = daily_base["calibration"]
    test = daily_base["test"]

    if calibration_window is None:
        calibration_window = CALIBRATION_WINDOW_DAYS.get(freq, 1)

    calibration_scores = conformity_scores(calibration, calibration_window)

    if freq != "D":
        test = test.resample(freq).sum()

    lower, upper, width, covered, alphas = adaptive_conformal_by_period(
        y_true=test["actual"].to_numpy(),
        y_pred=test["predicted"].to_numpy(),
        dates=test.index,
        calibration_scores=calibration_scores,
        alpha_target=alpha_target,
        gamma=gamma,
    )

    return test.assign(
        lower=lower,
        upper=upper,
        width=width,
        covered=covered,
        alpha_t=alphas,
    )


def build_group_monthly_conformal(
    predictions,
    group_column=GROUP_COL,
    alpha_target=0.10,
    gamma=0.01,
    calibration_window=None,
):
    """One monthly conformal frame per event group, keyed by group name."""
    if group_column not in predictions.columns:
        return {}

    test_records = (
        predictions[predictions["split"].eq("test")]
        .groupby(group_column, observed=True)
        .size()
    )

    group_conformal = {}
    for group_name in predictions[group_column].dropna().unique():
        monthly = conformalize_base(
            build_daily_base(
                predictions,
                group_column=group_column,
                group_name=group_name,
            ),
            freq="MS",
            alpha_target=alpha_target,
            gamma=gamma,
            calibration_window=calibration_window,
        )
        monthly.attrs["test_records"] = int(test_records.get(group_name, 0))
        group_conformal[group_name] = monthly

    return group_conformal
```

### Predictions

```python
###################################
# One prediction per target
###################################

SPLIT_ORDER = ["train", "calibration", "test"]
MODEL_LABEL = "tabular transformer"
POINT_METRIC_COLUMNS = ["n", "events", "D2", "RMSE", "MPD"]


def build_target_predictions(result, splits_raw, group_column=GROUP_COL):
    """One long frame per target: actual, predicted, split label and event group."""
    blocks = []
    for name in SPLIT_ORDER:
        frame = splits_raw[name]
        view = result["views"][name]
        block = pd.DataFrame(
            {"split": name,
             "actual": view["y"][:, 0].cpu().numpy(),
             "predicted": predict(result["params"], view)[:, 0]},
            index=frame.index)
        if group_column is not None and group_column in frame.columns:
            block[group_column] = frame[group_column].to_numpy()
        blocks.append(block)

    return pd.concat(blocks).sort_index(kind="stable")


def point_metrics(predictions):
    """n, events, D2, RMSE and MPD for one target, split by split."""
    rows = []
    for name in SPLIT_ORDER:
        part = predictions[predictions["split"].eq(name)]
        if len(part):
            rows.append({"split": name, **compute_metrics(part["actual"], part["predicted"])})

    return pd.DataFrame(rows).set_index("split")


def group_point_metrics(predictions, group_column=GROUP_COL):
    empty = pd.DataFrame(
        columns=POINT_METRIC_COLUMNS,
        index=pd.MultiIndex.from_arrays([[], []], names=["split", group_column]))

    if group_column not in predictions.columns:
        return empty

    rows = []
    for name in SPLIT_ORDER:
        part = predictions[predictions["split"].eq(name)]
        for group_name, block in part.groupby(group_column, observed=True):
            rows.append({"split": name, group_column: group_name,
                         **compute_metrics(block["actual"], block["predicted"])})

    return pd.DataFrame(rows).set_index(["split", group_column]) if rows else empty


def assemble_target_result(result, splits_raw, group_column=GROUP_COL,
                           alpha_target=0.10, gamma=0.01):
    predictions = build_target_predictions(result, splits_raw, group_column)
    daily_base = build_daily_base(predictions)

    return {
        **result,
        "model_name": MODEL_LABEL,
        "predictions": predictions,
        "metrics": point_metrics(predictions),
        "group_metrics": group_point_metrics(predictions, group_column),
        "daily_conformal": conformalize_base(daily_base, freq="D",
                                             alpha_target=alpha_target, gamma=gamma),
        "monthly_conformal": conformalize_base(daily_base, freq="MS",
                                               alpha_target=alpha_target, gamma=gamma),
        "group_monthly_conformal": build_group_monthly_conformal(
            predictions, group_column=group_column,
            alpha_target=alpha_target, gamma=gamma),
    }


def build_estimation_results(results, splits_raw, group_column=GROUP_COL,
                           alpha_target=0.10, gamma=0.01):
    estimation = {}

    for target, result in results.items():
        print(f"[{MODEL_LABEL}] scoring {target}...")
        estimation[target] = assemble_target_result(result, splits_raw, group_column,
                                                  alpha_target, gamma)

    return estimation

started = time.time()
estimation_results = build_estimation_results(results, splits_raw)
print(f"\nscoring wall time: {time.time() - started:.1f} s")
```

### Reporting

```python
#############################################################
# Reporting tables
#############################################################

def _order_splits(frame, column="Split"):
    """Train, validation, calibration, then test: the splits in the order they occur in time."""
    frame[column] = pd.Categorical(frame[column], categories=SPLIT_ORDER, ordered=True)
    return frame


def _select_targets(results, targets):
    """filter results per targets"""
    if targets is None:
        return results
    if isinstance(targets, str):
        targets = [targets]

    missing = [name for name in targets if name not in results]
    if missing:
        raise KeyError(f"{missing} not in results: available are {list(results)}")

    return {name: results[name] for name in targets}


def point_metrics_by_target(results, targets=None):
    """point metrics tables per outcome"""
    selected = _select_targets(results, targets)
    df = pd.concat({k: v["metrics"] for k, v in selected.items()})
    df.index.names = ["Target", "Split"]

    return _order_splits(df.reset_index()).set_index(["Target", "Split"]).sort_index()


def point_metrics_by_event_group(results, group_column=GROUP_COL, targets=None):
    """point metrics tables per outcome and event group"""
    frames = {
        k: v["group_metrics"]
        for k, v in _select_targets(results, targets).items()
        if v.get("group_metrics") is not None and not v["group_metrics"].empty
    }

    if not frames:
        index = pd.MultiIndex.from_arrays([[], [], []], names=["Target", "Split", group_column])
        return pd.DataFrame(columns=POINT_METRIC_COLUMNS, index=index)

    df = pd.concat(frames)
    df.index.names = ["Target", "Split", group_column]

    return _order_splits(df.reset_index()).set_index(["Target", "Split", group_column]).sort_index()


def event_monthly_conformal_summary(results, group_column=GROUP_COL):
    """Coverage per target and event group, read on the monthly view of the
    daily base."""
    rows = [
        {
            "Target": target,
            group_column: group_name,
            "Months": len(frame),
            "TestRecords": frame.attrs.get("test_records", 0),
            "Coverage": float(frame["covered"].mean()),
            "MeanActual": float(frame["actual"].mean()),
            "MeanWidth": float(frame["width"].mean()),
        }
        for target, result in results.items()
        for group_name, frame in result["group_monthly_conformal"].items()
    ]

    summary = pd.DataFrame(rows).set_index(["Target", group_column]).sort_index()

    return _add_relative_width(summary)


def _add_relative_width(summary):
    """Reads the band width against the level it is measuring.

    MeanWidth is in target units - deaths per month, injuries per day - so it
    cannot be compared across targets, views or event groups. RelWidth can:
    it is MeanWidth over MeanActual."""
    mean_actual = summary["MeanActual"]
    summary["RelWidth"] = summary["MeanWidth"] / mean_actual.where(mean_actual > 0)

    return summary


def daily_conformal_summary(results):
    """target coverage: it starts from daily results aggregated by target"""
    df = pd.concat({k: v["daily_conformal"] for k, v in results.items()})
    df.index.names = ["Target", "Date"]

    grouped = df.reset_index().groupby("Target")
    summary = pd.DataFrame({
        "Days": grouped["covered"].size(),
        "Coverage": grouped["covered"].mean(),
        "MeanActual": grouped["actual"].mean(),
        "MeanWidth": grouped["width"].mean(),
    })

    return _add_relative_width(summary)
```

### Historical backtesting

```python
###################################################
# Historical backtesting
###################################################



BACKTEST_VALIDATION_YEARS = (2019, 2020, 2021, 2022, 2023)


def expanding_year_splits(frame, validation_years=BACKTEST_VALIDATION_YEARS):
    """Train on every year strictly before val_year, validate on val_year alone."""
    years = frame.index.year.to_numpy()
    for fold, val_year in enumerate(validation_years, start=1):
        yield (fold, val_year,
               np.flatnonzero(years < val_year),
               np.flatnonzero(years == val_year))



def backtest_target_cv(frame, config, target, epochs,
                       validation_years=BACKTEST_VALIDATION_YEARS):
    """backtesting end-to-end pipeline."""
    index = PREDICTION_TARGETS.index(target)
    records = []

    print("=" * 78)
    print(f"backtesting {target} at {epochs} epochs per fold")
    print("=" * 78)

    for fold, val_year, train_index, val_index in expanding_year_splits(frame, validation_years):
        fold_train = frame.iloc[train_index].copy()
        fold_validation = frame.iloc[val_index].copy()
        encoders = fit_encoders(fold_train)
        tensors = {"train": encode_split(fold_train, encoders),
                   "validation": encode_split(fold_validation, encoders)}
        cardinalities = [len(encoders["vocabularies"][column]) + 1 for column in CATEGORICAL]
        views = {name: target_view(part, index) for name, part in tensors.items()}

        print(f"fold {fold}: train <= {val_year - 1} ({len(fold_train):,} rows)   "
              f"validation = {val_year} ({len(fold_validation):,} rows)")

        baseline_rate = (views["train"]["y"].sum(dim=0)
                         / views["train"]["exposure"].sum()).clamp(min=1e-8)
        fold_params = make_params(config, cardinalities, baseline_rate)
        train_model(fold_params, views["train"], config, epochs=epochs)
        predicted = predict(fold_params, views["validation"])[:, 0]

        records.append(pd.DataFrame(
            {"fold": fold, "validation_year": val_year,
             "actual": views["validation"]["y"][:, 0].cpu().numpy(),
             "predicted": predicted},
            index=fold_validation.index))

        del fold_params, tensors, views, fold_train, fold_validation
        gc.collect()
        torch.cuda.empty_cache()



    return pd.concat(records).sort_index(kind="stable")

```

## From casualties to injuries and deaths

The model predicts one number per tornado: `CASUALTY`, injuries plus deaths. This section checks whether that prediction can be split into its two parts **without building a second model**, and scores the split against the actual injuries and deaths.

```python
# ==============================================================================
# CASUALTY COMPONENTS, ALIGNED WITH MODEL SPLITS
# ==============================================================================

INJURY_PARTS = ["INJURIES_DIRECT", "INJURIES_INDIRECT"]
DEATH_PARTS = ["DEATHS_DIRECT", "DEATHS_INDIRECT"]
CASUALTY_PARTS = INJURY_PARTS + DEATH_PARTS


def build_casualty_splits(raw, model_splits):
    """Build injury/death components and align them with the model splits."""

    columns = ["BEGIN_DATE_TIME", "EVENT_GROUP", *CASUALTY_PARTS]
    components = raw[columns].copy()

    # Same row preparation used for the modelling dataset
    components["BEGIN_DATE_TIME"] = parse_storm_datetime(
        components["BEGIN_DATE_TIME"]
    )

    components = components.dropna(subset=["BEGIN_DATE_TIME"])
    components = components.loc[
        ~components["EVENT_GROUP"].isin(EXCLUDE_GROUPS)
    ]

    components.index = pd.DatetimeIndex(
        components["BEGIN_DATE_TIME"],
        name="Date"
    )

    components = components.sort_index(kind="stable")

    # Numerical casualty components
    components[CASUALTY_PARTS] = components[CASUALTY_PARTS].astype("float64")

    components["INJURIES"] = components[INJURY_PARTS].sum(axis=1)
    components["DEATHS"] = components[DEATH_PARTS].sum(axis=1)

    components = components.drop(
        columns=["BEGIN_DATE_TIME", "EVENT_GROUP"]
    )

    # Align with the exact modelling splits
    components_by_split = {}

    for split_name, model_frame in model_splits.items():

        part = components.loc[
            model_frame.index.min():model_frame.index.max()
        ]

        # Same rows and same order
        assert part.index.equals(model_frame.index), (
            f"{split_name}: casualty rows are not aligned"
        )

        # Injuries + deaths must reproduce CASUALTY exactly
        rebuilt_casualty = (
            part["INJURIES"] + part["DEATHS"]
        ).to_numpy()

        assert np.allclose(
            rebuilt_casualty,
            model_frame["CASUALTY"].to_numpy()
        ), f"{split_name}: casualty target mismatch"

        components_by_split[split_name] = part

    return components_by_split


components_by_split = build_casualty_splits(
    raw=df,
    model_splits=splits_raw
)

print(
    "Components rebuilt and aligned:",
    {name: len(part) for name, part in components_by_split.items()}
)
```

```python
# ==============================================================================
# INJURIES VS DEATHS, SPLIT BY SPLIT
# ==============================================================================

LARGE_EVENT_CASUALTIES = 100


def injury_death_summary(part):

    injuries = part["INJURIES"]
    deaths = part["DEATHS"]
    casualties = injuries + deaths

    without_large_events = casualties <= LARGE_EVENT_CASUALTIES
    total_casualties = casualties.sum()

    return {
        "events": len(part),

        "Pearson":
            injuries.corr(deaths),

        "Spearman":
            injuries.corr(deaths, method="spearman"),

        "Pearson without large events":
            injuries[without_large_events].corr(
                deaths[without_large_events]
            ),

        "P(deaths>0 | injuries>0)":
            deaths[injuries > 0].gt(0).mean(),

        "P(deaths>0 | injuries=0)":
            deaths[injuries == 0].gt(0).mean(),

        "death share":
            deaths.sum() / total_casualties
            if total_casualties > 0 else np.nan,
    }
```

```python
###########################################
# DEATHS BY INJURY SIZE
###########################################
INJURY_SIZE_BINS = [0, 5, 20, 100, np.inf]
INJURY_SIZE_LABELS = ["1-5", "6-20", "21-100", ">100"]


def deaths_by_injury_size(part):

    injured = part.loc[
        part["INJURIES"] > 0,
        ["INJURIES", "DEATHS"]
    ]

    injury_size = pd.cut(
        injured["INJURIES"],
        bins=INJURY_SIZE_BINS,
        labels=INJURY_SIZE_LABELS
    )

    grouped = injured.groupby(
        injury_size,
        observed=False
    )["DEATHS"]

    summary = grouped.agg(
        events="size",
        death_probability=lambda deaths: deaths.gt(0).mean(),
        mean_deaths="mean"
    )

    return summary.rename(
        columns={
            "death_probability": "share with at least one death",
            "mean_deaths": "mean deaths",
        }
    )
```

```python
# ==============================================================================
# DIRECT VS INDIRECT CASUALTIES, SPLIT BY SPLIT
# ==============================================================================

def direct_indirect_summary(part):

    casualty_parts = part[CASUALTY_PARTS]

    # Calculate once
    events = casualty_parts.gt(0).sum()
    totals = casualty_parts.sum()

    summary = {}

    for column in CASUALTY_PARTS:

        label = column.lower().replace("_", " ")

        summary[f"events with {label}"] = int(events[column])
        summary[f"total {label}"] = float(totals[column])

    total_injuries = totals[INJURY_PARTS].sum()
    total_deaths = totals[DEATH_PARTS].sum()

    summary["indirect share of injuries"] = (
        totals["INJURIES_INDIRECT"] / total_injuries
        if total_injuries > 0 else np.nan
    )

    summary["indirect share of deaths"] = (
        totals["DEATHS_INDIRECT"] / total_deaths
        if total_deaths > 0 else np.nan
    )

    return summary
```

#### How injuries and deaths move together

* **Pearson correlation**: agreement of the *values*. It is driven by the few large events.
* **Spearman correlation**: agreement of the *rankings*. It gives small and large events the same weight.
* **P(deaths > 0 | injuries > 0)**: among tornadoes that injured someone, the share that also killed someone.
* **Death share**: deaths as a fraction of all casualties.

A high Pearson with a low Spearman means injuries and deaths rise together in big events, while on small events deaths are rare and unrelated to the number injured.

### Step 1: Case selection and coherent information units

Failure and well-predicted cases come from the same meaningful 2025 positive-casualty pool and are selected by **absolute error**. Geography and event-narrative variables are changed as coherent blocks to avoid internally inconsistent counterfactuals.


```python
# ==============================================================================
# PREDICTION-FIRST INTERPRETABILITY SETTINGS
# ==============================================================================

EXPLAIN_TARGET = "CASUALTY"
EXPLAIN_SPLIT = "test"
PERMUTATION_SPLITS = ["calibration", "test"]
PERMUTATION_REPEATS = 5
EXPLAIN_SEED = DEFAULT_RANDOM_STATE
N_FAILURE_CASES = 3
N_SUCCESS_CASES = 3
SUCCESS_ACTUAL_QUANTILE = 0.75
TOP_INTERVENTIONS_PER_EVENT = 10
MATCHED_DONORS = 32
MATCH_LOWER_Q = 0.10
MATCH_UPPER_Q = 0.90
LOCAL_LABEL_WRAP = 46
DECOMP_TOP_DRIVERS = TOP_INTERVENTIONS_PER_EVENT
DECOMP_DONORS = 16
DECOMP_TOP_PAIR_TERMS = 4
DECOMP_PAIR_MIN_ABS_CASUALTIES = 0.05
DECOMP_PAIR_DETAIL_MIN_ABS_CASUALTIES = 0.01
DECOMP_ZERO_TOL = 1e-10

FEATURE_LABELS = {
    "STATE": "State",
    "CZ_TYPE": "County or zone type",
    "WFO": "Weather Forecast Office (WFO)",
    "CZ_TIMEZONE": "Time zone",
    "SOURCE": "Source reported the event",
    "DATA_SOURCE": "Data source (NOAA database era)",
    "EVENT_GROUP": "Event group",
    "risk": "Risk level (read from narrative)",
    "event_scope": "Event scope (read from narrative)",
    "DAY_DURATION": "Duration (days)",
    "DISTANCE": "Start-to-end distance (km)",
    "CTX_N_PREV_DAY": "Events earlier that day (nationwide)",
    "CTX_N_STATE_PREV_DAY": "Events earlier that day in the same state",
    "CTX_HOURS_SINCE_STATE_GROUP": "Hours since last same-group event in the state (cap 48)",
    "EPISODE_NARRATIVE": "Episode narrative (text_embedding)",
    "EVENT_NARRATIVE": "Event narrative (text_embedding)",
    "SEASONAL": "Month of the year",
    "EPOCH": "Year",
}

UNIT_LABELS = {
    "GEOGRAPHY_BLOCK": "Geography block",
    "EVENT_NARRATIVE_BLOCK": "Event narrative block",
}

COLOR_HELPS = "#2e8b57"
COLOR_HURTS = "#c43c39"
COLOR_NEUTRAL = "#8a8985"

FEATURE_TOKENS = TOKEN_NAMES[1:]
TOKEN_POSITION = {name: position for position, name in enumerate(TOKEN_NAMES)}


def feature_label(token_name):
    return FEATURE_LABELS.get(token_name, token_name)


def build_intervention_units():
    """Group only tokens that must move together for a coherent intervention."""
    grouped_specs = [
        ("GEOGRAPHY_BLOCK", ["STATE", "WFO", "CZ_TIMEZONE"]),
        ("EVENT_NARRATIVE_BLOCK", ["EVENT_NARRATIVE", "risk", "event_scope"]),
    ]

    member_to_group = {}
    group_members = {}
    for group_name, candidates in grouped_specs:
        present = [name for name in candidates if name in FEATURE_TOKENS]
        if len(present) >= 2:
            group_members[group_name] = present
            for name in present:
                member_to_group[name] = group_name

    units = {}
    emitted_groups = set()
    for token_name in FEATURE_TOKENS:
        group_name = member_to_group.get(token_name)
        if group_name is None:
            units[token_name] = [token_name]
        elif group_name not in emitted_groups:
            units[group_name] = group_members[group_name]
            emitted_groups.add(group_name)
    return units


INTERVENTION_UNITS = build_intervention_units()


def intervention_label(unit_name):
    if unit_name in UNIT_LABELS:
        members = INTERVENTION_UNITS[unit_name]
        member_text = " + ".join(feature_label(name) for name in members)
        return f"{UNIT_LABELS[unit_name]} ({member_text})"
    return feature_label(unit_name)


explain_params = results[EXPLAIN_TARGET]["params"]
explain_views = results[EXPLAIN_TARGET]["views"]
set_mode(explain_params, False)

event_frame = splits_raw[EXPLAIN_SPLIT]
train_match_frame = splits_raw["train"]

print(f"Target: {EXPLAIN_TARGET}")
print(f"Split explained: {EXPLAIN_SPLIT} ({len(event_frame):,} tornadoes)")
print(f"Training donor pool: {len(train_match_frame):,} historical tornadoes")
print(f"Matched donors per intervention: {MATCHED_DONORS}")
print(f"Feature tokens: {len(FEATURE_TOKENS)}")
print(f"Coherent intervention units: {len(INTERVENTION_UNITS)}")
for unit_name, members in INTERVENTION_UNITS.items():
    if len(members) > 1:
        print(f"  {unit_name}: {', '.join(members)}")
```

### Step 2: Same fitted Transformer

Every intervention is evaluated with the **same trained encoder, decoder and Poisson head**. No surrogate model is used.


```python
# ==============================================================================
# SAME TRAINED MODEL, STARTING FROM TOKEN CONTENT
# ==============================================================================

def token_content(params, batch):
    """Token content that depends on the tornado, excluding column identity."""
    with torch.no_grad():
        return build_tokens(params, batch) - params["column_identity"]


def log_expected_from_content(params, content, exposure, batch_size=PREDICT_BATCH_SIZE):
    """Log expected casualty count using the trained Transformer and Poisson head."""
    identity = params["column_identity"].detach()
    outputs = []
    with torch.no_grad():
        for start in range(0, len(content), batch_size):
            stop = min(start + batch_size, len(content))
            log_expected, _ = forward_from_tokens(
                params, content[start:stop] + identity, exposure[start:stop]
            )
            outputs.append(log_expected[:, 0])
    return torch.cat(outputs)


def expected_from_content(params, content, exposure, batch_size=PREDICT_BATCH_SIZE):
    return torch.exp(log_expected_from_content(params, content, exposure, batch_size))


check_batch = select_rows_from_split(
    explain_views[EXPLAIN_SPLIT], slice(0, min(256, len(event_frame)))
)
with torch.no_grad():
    direct_log_expected, _ = forward(explain_params, check_batch)
check_content = token_content(explain_params, check_batch)
rebuilt_log_expected = log_expected_from_content(
    explain_params, check_content, check_batch["exposure"]
)
max_gap = float((direct_log_expected[:, 0] - rebuilt_log_expected).abs().max())
print(f"Largest forward-pass difference: {max_gap:.2e} on the log scale")
print("OK" if max_gap < 1e-5 else "CHECK THE TOKEN FORWARD PASS")
```

### Step 3: Match plausible historical alternatives

Counterfactual values come from structurally similar historical tornadoes. The target is never used for matching; the tested information unit is allowed to vary.


```python
# ==============================================================================
# MATCHING CACHE: STRUCTURAL VARIABLES ONLY
# ==============================================================================

MATCH_NUMERIC = [
    "DAY_DURATION", "DISTANCE",
    "CTX_N_PREV_DAY", "CTX_N_STATE_PREV_DAY", "CTX_HOURS_SINCE_STATE_GROUP",
    "MONTH_SIN", "MONTH_COS", "YEAR",
]
MATCH_CATEGORICAL = [
    "STATE", "CZ_TYPE", "WFO", "CZ_TIMEZONE", "SOURCE", "TOR_F_SCALE",
]

TOKEN_MATCH_EXCLUSIONS = {
    "STATE": {"STATE", "WFO", "CZ_TIMEZONE"},
    "WFO": {"STATE", "WFO", "CZ_TIMEZONE"},
    "CZ_TIMEZONE": {"STATE", "WFO", "CZ_TIMEZONE"},
    "CZ_TYPE": {"CZ_TYPE"},
    "SOURCE": {"SOURCE"},
    "DATA_SOURCE": {"DATA_SOURCE"},
    "EVENT_GROUP": {"EVENT_GROUP"},
    "DAY_DURATION": {"DAY_DURATION"},
    "DISTANCE": {"DISTANCE"},
    "CTX_N_PREV_DAY": {"CTX_N_PREV_DAY"},
    "CTX_N_STATE_PREV_DAY": {"CTX_N_STATE_PREV_DAY"},
    "CTX_HOURS_SINCE_STATE_GROUP": {"CTX_HOURS_SINCE_STATE_GROUP"},
    "SEASONAL": {"MONTH_SIN", "MONTH_COS"},
    "EPOCH": {"YEAR"},
}

MATCH_WEIGHTS = {
    "EVENT_GROUP": 2.0, "STATE": 1.25, "CZ_TYPE": 0.75,
    "WFO": 0.50, "CZ_TIMEZONE": 0.50, "SOURCE": 0.50, "DATA_SOURCE": 0.50,
    "DAY_DURATION": 1.0, "DISTANCE": 1.0,
    "CTX_N_PREV_DAY": 0.75, "CTX_N_STATE_PREV_DAY": 0.75,
    "CTX_HOURS_SINCE_STATE_GROUP": 0.75,
    "MONTH_SIN": 0.75, "MONTH_COS": 0.75, "YEAR": 0.50,
}


def clean_category(series):
    return series.astype("string").fillna("__MISSING__").to_numpy(dtype=str)


def build_matching_cache(frame):
    cache = {"numeric": {}, "categorical": {}, "centre": {}, "scale": {}}
    for column in MATCH_NUMERIC:
        if column not in frame.columns:
            continue
        values = pd.to_numeric(frame[column], errors="coerce").to_numpy(dtype=float)
        centre = float(np.nanmedian(values))
        q25, q75 = np.nanpercentile(values, [25, 75])
        scale = float(q75 - q25)
        if not np.isfinite(scale) or scale < 1e-8:
            scale = float(np.nanstd(values))
        if not np.isfinite(scale) or scale < 1e-8:
            scale = 1.0
        filled = np.where(np.isfinite(values), values, centre)
        cache["numeric"][column] = (filled - centre) / scale
        cache["centre"][column] = centre
        cache["scale"][column] = scale
    for column in MATCH_CATEGORICAL:
        if column in frame.columns:
            cache["categorical"][column] = clean_category(frame[column])
    return cache


MATCH_CACHE = build_matching_cache(train_match_frame)


def unit_match_exclusions(unit_name):
    excluded = set()
    for token_name in INTERVENTION_UNITS[unit_name]:
        excluded.update(TOKEN_MATCH_EXCLUSIONS.get(token_name, {token_name}))
    return excluded


def matching_distance(row, unit_name):
    excluded = unit_match_exclusions(unit_name)
    distance = np.zeros(len(train_match_frame), dtype=float)
    total_weight = 0.0

    for column, train_values in MATCH_CACHE["numeric"].items():
        if column in excluded:
            continue
        value = pd.to_numeric(pd.Series([row[column]]), errors="coerce").iloc[0]
        if pd.isna(value):
            value = MATCH_CACHE["centre"][column]
        z = (float(value) - MATCH_CACHE["centre"][column]) / MATCH_CACHE["scale"][column]
        difference = np.clip(train_values - z, -4.0, 4.0)
        weight = MATCH_WEIGHTS.get(column, 1.0)
        distance += weight * difference ** 2
        total_weight += weight

    for column, train_values in MATCH_CACHE["categorical"].items():
        if column in excluded:
            continue
        value = str(row[column]) if pd.notna(row[column]) else "__MISSING__"
        weight = MATCH_WEIGHTS.get(column, 1.0)
        distance += weight * (train_values != value).astype(float)
        total_weight += weight

    if total_weight > 0:
        distance /= total_weight
    return distance


def matched_donor_indices(row, unit_name, n_donors=MATCHED_DONORS):
    distance = matching_distance(row, unit_name)
    n = min(n_donors, len(distance))
    candidate = np.argpartition(distance, n - 1)[:n]
    candidate = candidate[np.argsort(distance[candidate], kind="stable")]
    return candidate, distance[candidate]
```

### Step 4: Local prediction impact

**Question:** if one coherent information unit were replaced by plausible historical alternatives, how would the prediction change?  
`Delta > 0` means the original information raises the prediction; `Delta < 0` lowers it. `Helps fit / hurts fit` states whether keeping the original information moves the prediction closer to the observed count.


```python
# ==============================================================================
# COHERENT MATCHED LOCAL INTERVENTION + FAILURE / WELL-PREDICTED DIAGNOSTICS
# ==============================================================================

MONTH_NAMES = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
               "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]


def feature_value_text(row, token_name):
    if token_name in ("EPISODE_NARRATIVE", "EVENT_NARRATIVE"):
        return "text embedding"
    if token_name == "SEASONAL":
        return MONTH_NAMES[int(row["MONTH_BEGIN"]) - 1] if "MONTH_BEGIN" in row else "month"
    if token_name == "EPOCH":
        return str(int(row[EPOCH_COLUMN]))
    if token_name not in row.index:
        return ""
    value = row[token_name]
    if isinstance(value, (int, float, np.integer, np.floating)):
        return f"{value:,.3g}"
    return str(value)


def intervention_value_text(row, unit_name):
    members = INTERVENTION_UNITS[unit_name]
    if unit_name == "EVENT_NARRATIVE_BLOCK":
        details = []
        if "risk" in row.index:
            details.append(f"risk={row['risk']}")
        if "event_scope" in row.index:
            details.append(f"scope={row['event_scope']}")
        suffix = ", ".join(details)
        return "text" + (f"; {suffix}" if suffix else "")
    if unit_name == "GEOGRAPHY_BLOCK":
        details = [str(row[name]) for name in ("STATE", "WFO", "CZ_TIMEZONE") if name in row.index]
        return " / ".join(details)
    return feature_value_text(row, members[0])


def matched_intervention_effects(params, view, frame, position, n_donors=MATCHED_DONORS):
    event_batch = select_rows_from_split(view, slice(position, position + 1))
    event_content = token_content(params, event_batch)
    event_exposure = event_batch["exposure"]
    full_prediction = float(expected_from_content(params, event_content, event_exposure)[0])
    actual = float(event_batch["y"][0, 0].detach().cpu())
    full_absolute_error = abs(actual - full_prediction)
    row = frame.iloc[position]

    records = []
    donor_details = {}

    for unit_name, token_names in INTERVENTION_UNITS.items():
        donor_indices, donor_distances = matched_donor_indices(row, unit_name, n_donors)
        donor_positions = torch.as_tensor(donor_indices, dtype=torch.long, device=DEVICE)
        donor_batch = select_rows_from_split(explain_views["train"], donor_positions)
        donor_content = token_content(params, donor_batch)

        n = len(donor_indices)
        intervened = event_content.expand(n, -1, -1).clone()
        for token_name in token_names:
            token_position = TOKEN_POSITION[token_name]
            intervened[:, token_position] = donor_content[:, token_position]

        exposure = event_exposure.expand(n, -1)
        matched_prediction = expected_from_content(params, intervened, exposure).detach().cpu().numpy()
        impacts = full_prediction - matched_prediction
        median_matched_prediction = float(np.median(matched_prediction))
        matched_absolute_error = abs(actual - median_matched_prediction)
        fit_contribution = matched_absolute_error - full_absolute_error

        records.append({
            "unit": unit_name,
            "members": ", ".join(token_names),
            "full prediction": full_prediction,
            "median matched prediction": median_matched_prediction,
            "median impact": float(np.median(impacts)),
            "p10 impact": float(np.quantile(impacts, MATCH_LOWER_Q)),
            "p90 impact": float(np.quantile(impacts, MATCH_UPPER_Q)),
            "full absolute error": full_absolute_error,
            "matched absolute error": matched_absolute_error,
            "fit contribution": float(fit_contribution),
            "median donor distance": float(np.median(donor_distances)),
            "donors": n,
        })
        donor_details[unit_name] = {
            "indices": donor_indices,
            "distances": donor_distances,
            "matched_prediction": matched_prediction,
            "impact": impacts,
        }

    return {
        "position": position,
        "prediction": full_prediction,
        "actual": actual,
        "summary": pd.DataFrame(records).set_index("unit"),
        "donors": donor_details,
    }


def compact_intervention_label(unit_name):
    """Short plot label; the full unit definition is documented in Step 1."""
    return UNIT_LABELS.get(unit_name, feature_label(unit_name))


def selected_local_units(source, top=TOP_INTERVENTIONS_PER_EVENT):
    """The same local information units are reused by reconciliation, routing and Q/K/V."""
    summary = source["summary"] if isinstance(source, dict) else source
    return list(summary["median impact"].abs().sort_values(ascending=False).index[:top])


def fit_role(value, tolerance=1e-10):
    if value > tolerance:
        return "helps fit"
    if value < -tolerance:
        return "hurts fit"
    return "neutral for fit"


def plot_matched_intervention_effects(result, frame, case_kind, top=TOP_INTERVENTIONS_PER_EVENT):
    """One event -> one readable prediction-impact chart plus a separate interpretation column."""
    import textwrap
    from matplotlib.patches import Patch

    position = result["position"]
    row = frame.iloc[position]
    summary = result["summary"]

    units = selected_local_units(result, top=top)
    shown = summary.loc[units].sort_values("median impact")
    median = shown["median impact"].to_numpy()
    lower = shown["p10 impact"].to_numpy()
    upper = shown["p90 impact"].to_numpy()
    xerr = np.vstack([median - lower, upper - median])
    fit_values = shown["fit contribution"].to_numpy()

    labels = []
    for unit in shown.index:
        raw = f"{compact_intervention_label(unit)} = {intervention_value_text(row, unit)}"
        labels.append(textwrap.fill(raw, width=LOCAL_LABEL_WRAP, break_long_words=False))

    colors = [
        COLOR_HELPS if value > 1e-10 else COLOR_HURTS if value < -1e-10 else COLOR_NEUTRAL
        for value in fit_values
    ]
    positions = np.arange(len(shown), dtype=float)

    figure_height = max(7.2, 0.82 * len(shown) + 3.4)
    figure, (axis, notes) = plt.subplots(
        1, 2, figsize=(16.8, figure_height), sharey=True,
        gridspec_kw={"width_ratios": [3.5, 2.7], "wspace": 0.04},
    )

    axis.barh(positions, median, color=colors, height=0.58, alpha=0.90)
    axis.errorbar(
        median, positions, xerr=xerr, fmt="none", ecolor="#4f4f4f",
        elinewidth=1.4, capsize=3, zorder=3,
    )
    axis.axvline(0, color=COLOR_NEUTRAL, linewidth=1)
    axis.set_yticks(positions, labels)
    axis.tick_params(axis="y", labelsize=9.3, pad=8)
    axis.set_xlabel(
        "change in expected casualties from original information vs matched replacements\n"
        "bar = median; error line = 10th-90th percentile"
    )
    axis.spines[["top", "right"]].set_visible(False)
    axis.grid(axis="x", alpha=0.25)

    notes.set_xlim(0, 1)
    notes.set_xticks([])
    notes.tick_params(axis="y", left=False, labelleft=False)
    notes.spines[["top", "right", "bottom", "left"]].set_visible(False)
    notes.set_title("Impact and fit diagnosis", loc="left", fontsize=10.5)

    for y, (_, values) in zip(positions, shown.iterrows()):
        stability = (
            "stable direction"
            if (values["p10 impact"] > 0 and values["p90 impact"] > 0)
            or (values["p10 impact"] < 0 and values["p90 impact"] < 0)
            else "reference-sensitive"
        )
        role = fit_role(values["fit contribution"])
        if role == "helps fit":
            role_text = f"helps fit by {values['fit contribution']:.3f} abs-error units"
        elif role == "hurts fit":
            role_text = f"hurts fit by {abs(values['fit contribution']):.3f} abs-error units"
        else:
            role_text = "neutral for fit"

        notes.text(
            0.02, y,
            f"Delta {values['median impact']:+.3f}  "
            f"[{values['p10 impact']:+.3f}, {values['p90 impact']:+.3f}]\n"
            f"{role_text} | {stability}",
            va="center", ha="left", fontsize=9.0,
        )

    title_parts = [str(frame.index[position].date())]
    if "STATE" in row.index:
        title_parts.append(str(row["STATE"]))
    figure.suptitle(
        f"{case_kind.upper()} CASE — {'  |  '.join(title_parts)}\n"
        f"predicted {result['prediction']:.3f} expected casualties | actual {result['actual']:.0f}",
        x=0.09, ha="left", y=0.985, fontsize=12,
    )
    axis.legend(
        handles=[
            Patch(color=COLOR_HELPS, label="original information helps fit observed count"),
            Patch(color=COLOR_HURTS, label="original information hurts fit observed count"),
        ],
        frameon=False, loc="lower right", fontsize=8.8,
    )
    figure.subplots_adjust(left=0.28, right=0.98, top=0.88, bottom=0.13)
    plt.show()


# ------------------------------------------------------------------------------
# Select failure and well-predicted cases from the same meaningful 2025 pool
# ------------------------------------------------------------------------------
test_observed = explain_views[EXPLAIN_SPLIT]["y"][:, 0].detach().cpu().numpy().astype(float)
test_predicted = predict(explain_params, explain_views[EXPLAIN_SPLIT])[:, 0].astype(float)
test_absolute_error = np.abs(test_observed - test_predicted)

positive_observed = test_observed[test_observed > 0]
meaningful_actual_threshold = (
    max(1.0, float(np.quantile(positive_observed, SUCCESS_ACTUAL_QUANTILE)))
    if len(positive_observed) else 1.0
)

eligible_mask = test_observed >= meaningful_actual_threshold
eligible_positions = np.flatnonzero(eligible_mask)

# Fallback only if the upper positive tail is too small for both groups.
if len(eligible_positions) < N_FAILURE_CASES + N_SUCCESS_CASES:
    eligible_positions = np.flatnonzero(test_observed > 0)

failure_order = np.argsort(-test_absolute_error[eligible_positions], kind="stable")
failure_positions = eligible_positions[failure_order[:N_FAILURE_CASES]].astype(int)

remaining_candidates = eligible_positions[~np.isin(eligible_positions, failure_positions)]
success_order = np.argsort(test_absolute_error[remaining_candidates], kind="stable")
success_positions = remaining_candidates[success_order[:N_SUCCESS_CASES]].astype(int)

CASE_KIND = {int(position): "Failure" for position in failure_positions}
CASE_KIND.update({int(position): "Well-predicted" for position in success_positions})
selected_positions = np.concatenate([failure_positions, success_positions]).astype(int)


def case_selection_table(positions_group):
    positions_group = np.asarray(positions_group, dtype=int)
    print(f"{len(positions_group)} cases selected")
    return pd.DataFrame({
        "date": event_frame.index[positions_group],
        "state": (event_frame["STATE"].to_numpy()[positions_group]
                  if "STATE" in event_frame.columns else ""),
        "actual": test_observed[positions_group],
        "predicted": test_predicted[positions_group],
        "absolute error": test_absolute_error[positions_group],
    }).set_index("date").round(3)


def run_local_cases(case_kind, positions_group):
    """Table + one impact chart per event for a single case group."""
    print(f"===== {case_kind.upper()} CASES =====")
    display(case_selection_table(positions_group))
    for position in positions_group:
        position = int(position)
        result = matched_intervention_effects(
            explain_params, explain_views[EXPLAIN_SPLIT], event_frame, position
        )
        result["case_kind"] = case_kind
        MATCHED_LOCAL[position] = result
        plot_matched_intervention_effects(result, event_frame, case_kind)

MATCHED_LOCAL = {}

```

### Step 5: Exact prediction reconciliation

**Question:** how does the Transformer move from a matched historical baseline to the exact fitted prediction?  
The waterfall reconciles **baseline + main effects + interactions = prediction**. The title separately reports `actual - prediction`, which is model error, not a feature contribution.


```python
# ==============================================================================
# EXACT LOCAL PREDICTION RECONCILIATION IN EXPECTED CASUALTIES
# ==============================================================================

from itertools import combinations

REMAINING_INFORMATION = "REMAINING_INFORMATION"


def matching_distance_with_exclusions(row, excluded):
    """Same structural matcher as Step 3, with an explicit exclusion set."""
    distance = np.zeros(len(train_match_frame), dtype=float)
    total_weight = 0.0

    for column, train_values in MATCH_CACHE["numeric"].items():
        if column in excluded:
            continue
        value = pd.to_numeric(pd.Series([row[column]]), errors="coerce").iloc[0]
        if pd.isna(value):
            value = MATCH_CACHE["centre"][column]
        z = (float(value) - MATCH_CACHE["centre"][column]) / MATCH_CACHE["scale"][column]
        difference = np.clip(train_values - z, -4.0, 4.0)
        weight = MATCH_WEIGHTS.get(column, 1.0)
        distance += weight * difference ** 2
        total_weight += weight

    for column, train_values in MATCH_CACHE["categorical"].items():
        if column in excluded:
            continue
        value = str(row[column]) if pd.notna(row[column]) else "__MISSING__"
        weight = MATCH_WEIGHTS.get(column, 1.0)
        distance += weight * (train_values != value).astype(float)
        total_weight += weight

    if total_weight > 0:
        distance /= total_weight
    return distance


def decomposition_groups(local_result, top_drivers=DECOMP_TOP_DRIVERS):
    """Local displayed units stay explicit; all non-selected units form one documented joint block."""
    selected_units = selected_local_units(local_result, top=top_drivers)
    groups = {unit_name: list(INTERVENTION_UNITS[unit_name]) for unit_name in selected_units}

    remaining_units = [name for name in INTERVENTION_UNITS if name not in selected_units]
    remaining_tokens = []
    for unit_name in remaining_units:
        remaining_tokens.extend(INTERVENTION_UNITS[unit_name])
    if remaining_tokens:
        groups[REMAINING_INFORMATION] = remaining_tokens

    return groups, selected_units, remaining_units


def decomposition_donor_indices(row, selected_units, n_donors=DECOMP_DONORS):
    """Real historical references matched while allowing the selected local drivers to vary."""
    excluded = set()
    for unit_name in selected_units:
        excluded.update(unit_match_exclusions(unit_name))

    distance = matching_distance_with_exclusions(row, excluded)
    n = min(n_donors, len(distance))
    candidate = np.argpartition(distance, n - 1)[:n]
    candidate = candidate[np.argsort(distance[candidate], kind="stable")]
    return candidate, distance[candidate]


def coalition_predictions(params, view, frame, position, groups, donor_indices):
    """Evaluate every coalition from real-donor content to the original event content."""
    event_batch = select_rows_from_split(view, slice(position, position + 1))
    event_content = token_content(params, event_batch)
    event_exposure = event_batch["exposure"]

    donor_positions = torch.as_tensor(donor_indices, dtype=torch.long, device=DEVICE)
    donor_batch = select_rows_from_split(explain_views["train"], donor_positions)
    donor_content = token_content(params, donor_batch)

    group_names = list(groups)
    n_groups = len(group_names)
    n_coalitions = 1 << n_groups
    n_donors = len(donor_indices)

    hybrid = donor_content[:, None, :, :].expand(
        n_donors, n_coalitions, donor_content.shape[1], donor_content.shape[2]
    ).clone()

    for group_index, group_name in enumerate(group_names):
        token_positions = [TOKEN_POSITION[name] for name in groups[group_name]]
        masks = [mask for mask in range(n_coalitions) if mask & (1 << group_index)]
        for mask in masks:
            hybrid[:, mask, token_positions, :] = event_content[0, token_positions, :]

    flat = hybrid.reshape(n_donors * n_coalitions, hybrid.shape[-2], hybrid.shape[-1])
    exposure = event_exposure.expand(n_donors * n_coalitions, -1)
    predictions = expected_from_content(params, flat, exposure).detach().cpu().numpy()
    predictions = predictions.reshape(n_donors, n_coalitions)
    return predictions, group_names


def mobius_dividends(coalition_values, n_groups):
    """Möbius inversion: exact Harsanyi dividends for every donor and coalition."""
    dividends = np.asarray(coalition_values, dtype=float).copy()
    n_coalitions = 1 << n_groups
    for bit in range(n_groups):
        for mask in range(n_coalitions):
            if mask & (1 << bit):
                dividends[:, mask] -= dividends[:, mask ^ (1 << bit)]
    return dividends


def group_display_label(group_name):
    if group_name == REMAINING_INFORMATION:
        return "Remaining model information (joint)"
    return compact_intervention_label(group_name)


def exact_prediction_reconciliation(params, view, frame, position, local_result,
                                    top_drivers=DECOMP_TOP_DRIVERS,
                                    n_donors=DECOMP_DONORS):
    """Exact accounting: baseline + main + pair + higher-order = fitted prediction."""
    row = frame.iloc[position]
    groups, selected_units, remaining_units = decomposition_groups(
        local_result, top_drivers=top_drivers
    )
    donor_indices, donor_distances = decomposition_donor_indices(
        row, selected_units, n_donors=n_donors
    )

    predictions, group_names = coalition_predictions(
        params, view, frame, position, groups, donor_indices
    )
    n_groups = len(group_names)
    n_coalitions = 1 << n_groups
    dividends = mobius_dividends(predictions, n_groups)
    mean_dividend = dividends.mean(axis=0)

    full_mask = n_coalitions - 1
    prediction = float(predictions[:, full_mask].mean())
    actual = float(local_result["actual"])
    residual = actual - prediction
    baseline = float(mean_dividend[0])

    main_terms = []
    for index, group_name in enumerate(group_names):
        mask = 1 << index
        main_terms.append({
            "kind": "main",
            "name": group_display_label(group_name),
            "raw_name": group_name,
            "value": float(mean_dividend[mask]),
            "p10": float(np.quantile(dividends[:, mask], MATCH_LOWER_Q)),
            "p90": float(np.quantile(dividends[:, mask], MATCH_UPPER_Q)),
        })

    pair_terms = []
    for left, right in combinations(range(n_groups), 2):
        mask = (1 << left) | (1 << right)
        pair_terms.append({
            "kind": "pair",
            "name": f"{group_display_label(group_names[left])} × {group_display_label(group_names[right])}",
            "raw_names": (group_names[left], group_names[right]),
            "value": float(mean_dividend[mask]),
            "p10": float(np.quantile(dividends[:, mask], MATCH_LOWER_Q)),
            "p90": float(np.quantile(dividends[:, mask], MATCH_UPPER_Q)),
        })

    higher_order_by_order = {}
    for interaction_order in range(3, n_groups + 1):
        masks = [
            mask for mask in range(1, n_coalitions)
            if int(mask).bit_count() == interaction_order
        ]
        if masks:
            higher_order_by_order[interaction_order] = float(mean_dividend[masks].sum())
    higher_order = float(sum(higher_order_by_order.values()))

    candidate_pairs = [
        item for item in pair_terms
        if REMAINING_INFORMATION not in item["raw_names"]
        and abs(item["value"]) >= DECOMP_PAIR_MIN_ABS_CASUALTIES
    ]
    candidate_pairs = sorted(candidate_pairs, key=lambda item: abs(item["value"]), reverse=True)
    shown_pairs = candidate_pairs[:DECOMP_TOP_PAIR_TERMS]
    shown_pair_names = {item["raw_names"] for item in shown_pairs}
    remaining_pair_terms = [
        item for item in pair_terms if item["raw_names"] not in shown_pair_names
    ]
    remaining_pair_value = float(sum(item["value"] for item in remaining_pair_terms))

    components = list(main_terms)
    components.extend(shown_pairs)
    if abs(remaining_pair_value) > DECOMP_ZERO_TOL:
        components.append({
            "kind": "pair_remainder",
            "name": "Remaining pair interactions",
            "value": remaining_pair_value,
        })
    if abs(higher_order) > DECOMP_ZERO_TOL:
        components.append({
            "kind": "higher",
            "name": "Interactions of order 3+",
            "value": higher_order,
        })

    reconstructed = baseline + sum(item["value"] for item in components)
    reconciliation_gap = prediction - reconstructed
    if abs(reconciliation_gap) > DECOMP_ZERO_TOL:
        components.append({
            "kind": "numerical",
            "name": "Numerical closure",
            "value": float(reconciliation_gap),
        })
        reconstructed += reconciliation_gap

    if abs(prediction - reconstructed) > 1e-6:
        raise RuntimeError("Prediction reconciliation did not close numerically.")

    remaining_main_value = 0.0
    for item in main_terms:
        if item.get("raw_name") == REMAINING_INFORMATION:
            remaining_main_value = float(item["value"])
            break

    return {
        "position": int(position),
        "actual": actual,
        "prediction": prediction,
        "residual": residual,
        "baseline": baseline,
        "components": components,
        "main_terms": main_terms,
        "pair_terms": pair_terms,
        "shown_pairs": shown_pairs,
        "remaining_pair_terms": remaining_pair_terms,
        "remaining_pair_value": remaining_pair_value,
        "higher_order": higher_order,
        "higher_order_by_order": higher_order_by_order,
        "reconstructed": float(reconstructed),
        "donor_indices": donor_indices,
        "donor_distances": donor_distances,
        "baseline_predictions": predictions[:, 0],
        "groups": groups,
        "selected_units": selected_units,
        "remaining_units": remaining_units,
        "remaining_main_value": remaining_main_value,
    }


def plot_prediction_reconciliation(result, frame):
    """One event -> one clearly cumulative waterfall chart."""
    import textwrap

    position = result["position"]
    row = frame.iloc[position]
    case_kind = CASE_KIND.get(position, "Case")

    step_items = [{
        "kind": "baseline",
        "name": "Matched historical baseline",
        "value": result["baseline"],
    }] + list(result["components"])

    display_items = list(step_items) + [{
        "kind": "total",
        "name": "Transformer prediction",
        "value": result["prediction"],
    }]

    figure_height = max(8.6, 0.60 * len(display_items) + 2.5)
    figure, axis = plt.subplots(figsize=(14.4, figure_height))

    type_colors = {
        "baseline": "#6f7782",
        "main": "#3f6f9f",
        "pair": "#7b5aa6",
        "pair_remainder": "#8e7ca8",
        "higher": "#b1783f",
        "numerical": "#8a8985",
        "total": "#2f5d50",
    }

    bars = []
    running = 0.0
    cumulative_after = []
    for item in step_items:
        start = running
        end = running + float(item["value"])
        bars.append((start, end, item))
        running = end
        cumulative_after.append(running)

    y_positions = np.arange(len(display_items))
    labels = []

    all_x = [0.0, result["prediction"]]
    for start, end, _ in bars:
        all_x.extend([start, end])
    x_min = min(all_x)
    x_max = max(all_x)
    span = max(x_max - x_min, 1e-6)
    text_pad = 0.018 * span
    annotation_x = x_max + 0.04 * span

    for y, ((start, end, item), cumulative) in enumerate(zip(bars, cumulative_after)):
        left = min(start, end)
        width = abs(end - start)
        axis.barh(
            y, width, left=left, height=0.62,
            color=type_colors.get(item["kind"], "#8a8985"), alpha=0.92,
            edgecolor="white", linewidth=1.0,
        )
        axis.plot([start, end], [y, y], color="white", linewidth=1.0, alpha=0.65)
        if y < len(step_items) - 1:
            axis.plot([end, end], [y + 0.31, y + 0.69], color="#555555", linewidth=0.9, alpha=0.55)

        delta_text = f"{float(item['value']):+.3f} → {cumulative:.3f}"
        axis.text(
            annotation_x, y, delta_text,
            va="center", ha="left", fontsize=9.2, fontweight="bold"
        )
        labels.append(textwrap.fill(item["name"], width=38, break_long_words=False))

    total_y = len(display_items) - 1
    total_value = float(result["prediction"])
    axis.barh(
        total_y, total_value, left=0.0, height=0.62,
        color=type_colors["total"], alpha=0.96, edgecolor="white", linewidth=1.0,
    )
    axis.text(
        annotation_x, total_y, f"total {total_value:.3f}",
        va="center", ha="left", fontsize=9.4, fontweight="bold",
    )
    labels.append("Transformer prediction")

    axis.axvline(0.0, color="black", linewidth=0.9, alpha=0.65)
    axis.set_yticks(y_positions, labels)
    axis.invert_yaxis()
    axis.set_xlabel("expected casualties — exact reconciliation of the fitted prediction")
    axis.set_ylabel("")
    axis.grid(axis="x", alpha=0.16)

    title_parts = [str(frame.index[position].date())]
    if "STATE" in row.index:
        title_parts.append(str(row["STATE"]))
    axis.set_title(
        f"{case_kind.upper()} CASE — prediction reconciliation — {' | ' .join(title_parts)}\n"
        f"prediction {result['prediction']:.3f} | actual {result['actual']:.3f} | "
        f"unexplained residual {result['residual']:+.3f}",
        loc="left", pad=12,
    )

    axis.set_xlim(x_min - 0.10 * span, x_max + 0.34 * span)
    figure.subplots_adjust(left=0.44, right=0.97, top=0.90, bottom=0.08)
    plt.show()


def display_reconciliation_detail(result):
    """Only concise two-column tables: variable + contribution."""

    if result["remaining_units"]:
        variable_text = "Remaining model information (joint): " + ", ".join(
            compact_intervention_label(unit_name) for unit_name in result["remaining_units"]
        )
        remaining_df = pd.DataFrame([
            {"variable": variable_text, "contribution": result["remaining_main_value"]}
        ]).round({"contribution": 3})
        print("Aggregated main-effect detail:")
        display(remaining_df)

    if result["remaining_pair_terms"]:
        pair_rows = [
            {"variable": item["name"], "contribution": item["value"]}
            for item in sorted(result["remaining_pair_terms"], key=lambda x: abs(x["value"]), reverse=True)
            if abs(item["value"]) > DECOMP_ZERO_TOL
        ]
        if pair_rows:
            print("Remaining pair-interaction detail:")
            display(pd.DataFrame(pair_rows).round({"contribution": 3}))

    if result["higher_order_by_order"]:
        higher_rows = [
            {"variable": f"{order}-way interactions", "contribution": value}
            for order, value in sorted(result["higher_order_by_order"].items())
            if abs(value) > DECOMP_ZERO_TOL
        ]
        if higher_rows:
            print("Higher-order interaction detail:")
            display(pd.DataFrame(higher_rows).round({"contribution": 3}))



def run_reconciliation_cases(case_kind, positions_group):
    """Waterfall chart + detail tables for every event of a single case group."""
    print(f"===== {case_kind.upper()} CASES: PREDICTION RECONCILIATION =====")
    for position in positions_group:
        position = int(position)
        reconciliation = exact_prediction_reconciliation(
            explain_params,
            explain_views[EXPLAIN_SPLIT],
            event_frame,
            position,
            MATCHED_LOCAL[position],
        )
        PREDICTION_RECONCILIATION[position] = reconciliation
        plot_prediction_reconciliation(reconciliation, event_frame)
        display_reconciliation_detail(reconciliation)


PREDICTION_RECONCILIATION = {}
```

### Step 6: Internal routing

**Question:** where does already-verified information travel inside the Transformer?  
The same local drivers are traced through **Encoder → CLS**, **Decoder attention**, and **Attention × gradient**. Shares are labelled `dominant`, `moderate` or `weak`; routing is not casualty contribution.


```python
# ==============================================================================
# EXPLAINABLE ATTENTION PASS: SAME TRAINED WEIGHTS, PER-HEAD MAPS KEPT IN GRAPH
# ==============================================================================

ROUTING_TOP_DRIVERS = TOP_INTERVENTIONS_PER_EVENT
ROUTING_EPS = 1e-12


def attention_with_head_weights(layer, query, key, value):
    """Reproduce nn.MultiheadAttention in evaluation mode and retain per-head weights."""
    d_model = layer.embed_dim
    n_heads = layer.num_heads
    head_width = d_model // n_heads

    weight_q, weight_k, weight_v = layer.in_proj_weight.chunk(3, dim=0)
    bias_q, bias_k, bias_v = layer.in_proj_bias.chunk(3, dim=0)

    def split_heads(values):
        rows, length, _ = values.shape
        return values.view(rows, length, n_heads, head_width).transpose(1, 2)

    queries = split_heads(F.linear(query, weight_q, bias_q))
    keys = split_heads(F.linear(key, weight_k, bias_k))
    values = split_heads(F.linear(value, weight_v, bias_v))

    scores = queries @ keys.transpose(-2, -1) / math.sqrt(head_width)
    weights = torch.softmax(scores, dim=-1)
    mixed = (weights @ values).transpose(1, 2).reshape(query.shape[0], query.shape[1], d_model)
    return layer.out_proj(mixed), weights


def forward_with_routing_maps(params, tokens, exposure):
    """Same encoder -> decoder -> Poisson head path, returning per-head attention maps."""
    encoder_maps = []

    for block in params["encoder_blocks"]:
        normalized = block["norm_attention"](tokens)
        attention_output, weights = attention_with_head_weights(
            block["attention"], normalized, normalized, normalized
        )
        tokens = tokens + attention_output
        encoder_maps.append(weights)

        normalized = block["norm_ffn"](tokens)
        tokens = tokens + block["ffn_out"](F.gelu(block["ffn_in"](normalized)))

    memory = params["encoder_norm_out"](tokens)

    queries = memory[:, :1, :]
    feature_memory = memory[:, 1:, :]
    attention_output, decoder_map = attention_with_head_weights(
        params["decoder_attention"], queries, feature_memory, feature_memory
    )
    queries = queries + attention_output
    normalized = params["decoder_norm_ffn"](queries)
    queries = queries + params["decoder_ffn_out"](
        F.gelu(params["decoder_ffn_in"](normalized))
    )
    queries = params["decoder_norm_out"](queries)

    log_expected = apply_head(params, queries, exposure)
    return log_expected, encoder_maps, decoder_map


routing_check_batch = select_rows_from_split(
    explain_views[EXPLAIN_SPLIT], slice(0, min(128, len(event_frame)))
)
with torch.no_grad():
    reference_log_expected, _ = forward(explain_params, routing_check_batch)
    routing_log_expected, _, _ = forward_with_routing_maps(
        explain_params,
        build_tokens(explain_params, routing_check_batch),
        routing_check_batch["exposure"],
    )
routing_gap = float((reference_log_expected - routing_log_expected).abs().max())
print(f"Largest model vs routing-pass difference: {routing_gap:.2e} on the log scale")
print("OK" if routing_gap < 1e-5 else "CHECK THE EXPLAINABLE ATTENTION PASS")
```

### Reading the routing bars

`dominant ≥ 15%`, `moderate = 5–15%`, `weak < 5%`. Drivers are ordered from dominant to weak using their strongest share across the three routing stages.


```python
# ==============================================================================
# ROUTING METRICS FOR THE SAME LOCAL INFORMATION UNITS
# ==============================================================================


def normalize_feature_share(values):
    """Normalize non-negative feature values to shares summing to one."""
    values = values.clamp(min=0)
    return values / (values.sum(dim=-1, keepdim=True) + ROUTING_EPS)


def encoder_rollout_to_cls(encoder_maps):
    """Residual-aware attention rollout from original tokens to the final encoded CLS."""
    rows, _, n_tokens, _ = encoder_maps[0].shape
    eye = torch.eye(n_tokens, device=encoder_maps[0].device).expand(rows, -1, -1)
    rollout = eye

    for weights in encoder_maps:
        mean_attention = weights.mean(dim=1)
        layer_route = 0.5 * eye + 0.5 * mean_attention
        rollout = layer_route @ rollout

    cls_sources = rollout[:, 0, 1:]
    return normalize_feature_share(cls_sources)


def decoder_attention_share(decoder_map):
    share = decoder_map.mean(dim=1)[:, 0, :]
    return normalize_feature_share(share)


def gradient_decoder_attention_share(decoder_map, decoder_gradient):
    relevance = (decoder_map * decoder_gradient).abs().mean(dim=1)[:, 0, :]
    return normalize_feature_share(relevance)


def aggregate_routing_by_unit(feature_values):
    """Sum token-level routing shares into the coherent intervention units."""
    aggregated = {}
    for unit_name, token_names in INTERVENTION_UNITS.items():
        positions = [FEATURE_TOKENS.index(name) for name in token_names]
        aggregated[unit_name] = float(feature_values[positions].sum())
    return aggregated


def uniform_share_for_unit(unit_name):
    """Expected unit share under uniform routing across feature tokens."""
    return len(INTERVENTION_UNITS[unit_name]) / max(len(FEATURE_TOKENS), 1)


def routing_audit_for_event(params, view, position, local_result):
    batch = select_rows_from_split(view, slice(position, position + 1))
    tokens = build_tokens(params, batch)
    exposure = batch["exposure"]

    log_expected, encoder_maps, decoder_map = forward_with_routing_maps(params, tokens, exposure)
    decoder_gradient = torch.autograd.grad(log_expected.sum(), decoder_map)[0]

    encoder_share = encoder_rollout_to_cls(encoder_maps)[0]
    decoder_share = decoder_attention_share(decoder_map)[0]
    gradient_share = gradient_decoder_attention_share(decoder_map, decoder_gradient)[0]

    encoder_units = aggregate_routing_by_unit(encoder_share)
    decoder_units = aggregate_routing_by_unit(decoder_share)
    gradient_units = aggregate_routing_by_unit(gradient_share)

    local_summary = local_result["summary"].copy()
    local_summary["encoder_to_cls"] = pd.Series(encoder_units)
    local_summary["decoder_attention"] = pd.Series(decoder_units)
    local_summary["gradient_decoder_attention"] = pd.Series(gradient_units)
    local_summary["robust_direction"] = (
        ((local_summary["p10 impact"] > 0) & (local_summary["p90 impact"] > 0))
        | ((local_summary["p10 impact"] < 0) & (local_summary["p90 impact"] < 0))
    )

    return {
        "position": position,
        "prediction": local_result["prediction"],
        "actual": local_result["actual"],
        "summary": local_summary,
    }




def qualitative_share_label(share):
    """Simple descriptive band for readability."""
    if share >= 0.15:
        return "dominant"
    if share >= 0.05:
        return "moderate"
    return "weak"


def plot_share_panels(result, frame, value_columns, panel_titles, main_title, top=ROUTING_TOP_DRIVERS):
    """Readable routing/QKV display: one horizontal bar panel per metric."""
    import textwrap

    position = result["position"]
    row = frame.iloc[position]
    summary = result["summary"]

    drivers = selected_local_units(summary, top=top)
    candidate = summary.loc[drivers].copy()
    max_share = candidate[value_columns].max(axis=1)
    category_rank = max_share.apply(lambda x: 0 if x >= 0.15 else (1 if x >= 0.05 else 2))
    ordering = pd.DataFrame({"rank": category_rank, "share": max_share}).sort_values(
        ["rank", "share"], ascending=[True, False], kind="stable"
    ).index
    shown = candidate.loc[ordering]

    labels = []
    for unit_name, values in shown.iterrows():
        stability = "stable" if values["robust_direction"] else "reference-sensitive"
        role = fit_role(values["fit contribution"])
        base = compact_intervention_label(unit_name)
        effect = (
            f"Delta {values['median impact']:+.3f} "
            f"[{values['p10 impact']:+.3f}, {values['p90 impact']:+.3f}] | {role} | {stability}"
        )
        labels.append(textwrap.fill(base, width=36, break_long_words=False) + "\n" + effect)

    values_matrix = 100.0 * shown[value_columns].to_numpy(dtype=float)
    x_max = max(25.0, float(np.nanmax(values_matrix)) * 1.28)

    figure_height = max(7.4, 1.18 * len(shown) + 2.8)
    figure, axes = plt.subplots(
        1, len(value_columns), figsize=(16.8, figure_height), sharey=True,
        gridspec_kw={"wspace": 0.08}
    )
    if len(value_columns) == 1:
        axes = [axes]

    y_positions = np.arange(len(labels))
    panel_colors = ["#3f6f9f", "#5f8bb6", "#7aa6cf"]

    for panel_idx, (axis, column, title) in enumerate(zip(axes, value_columns, panel_titles)):
        panel_values = values_matrix[:, panel_idx]
        axis.barh(y_positions, panel_values, color=panel_colors[panel_idx % len(panel_colors)], alpha=0.92)
        axis.set_xlim(0, x_max)
        axis.set_title(title, fontsize=11, pad=10)
        axis.grid(axis="x", alpha=0.16)
        axis.invert_yaxis()
        axis.set_xlabel("share (%)")
        axis.tick_params(axis="x", labelsize=9)
        if panel_idx == 0:
            axis.set_yticks(y_positions, labels)
            axis.tick_params(axis="y", labelsize=9.1, pad=8)
        else:
            axis.tick_params(axis="y", left=False, labelleft=False)

        for y, share in enumerate(panel_values):
            descriptor = qualitative_share_label(share / 100.0)
            axis.text(
                x_max * 0.985, y, f"{share:.1f}%   {descriptor}",
                va="center", ha="right", fontsize=9.2, fontweight="bold",
                bbox={"boxstyle": "round,pad=0.16", "facecolor": "white",
                      "alpha": 0.85, "edgecolor": "none"},
            )

    title_parts = [str(frame.index[position].date())]
    if "STATE" in row.index:
        title_parts.append(str(row["STATE"]))
    case_kind = CASE_KIND.get(int(position), "Case")
    axes[0].set_title(panel_titles[0], fontsize=11, pad=10)
    figure.suptitle(
        f"{case_kind.upper()} CASE — {main_title} — {' | ' .join(title_parts)}\n"
        f"predicted {result['prediction']:.3f} expected casualties | actual {result['actual']:.0f}",
        x=0.31, y=0.98, ha="left", fontsize=12
    )
    figure.subplots_adjust(left=0.47, right=0.98, top=0.84, bottom=0.08)
    plt.show()


def plot_routing_audit(result, frame, top=ROUTING_TOP_DRIVERS):
    plot_share_panels(
        result,
        frame,
        value_columns=["encoder_to_cls", "decoder_attention", "gradient_decoder_attention"],
        panel_titles=["Encoder -> CLS", "Decoder attention", "Attention x gradient"],
        main_title="internal routing",
        top=top,
    )


def run_routing_cases(case_kind, positions_group):
    """Internal-routing chart for every event of a single case group."""
    print(f"===== {case_kind.upper()} CASES: INTERNAL ROUTING =====")
    for position in positions_group:
        position = int(position)
        ROUTING_AUDIT[position] = routing_audit_for_event(
            explain_params,
            explain_views[EXPLAIN_SPLIT],
            position,
            MATCHED_LOCAL[position],
        )
        plot_routing_audit(ROUTING_AUDIT[position], event_frame)


ROUTING_AUDIT = {}
```

### Step 7: Q/K/V mechanism

**Question:** why does the decoder select and transmit particular information?  
**Q–K selection** is what the CLS query selects, **V payload** is available content, and **Delivered V** is what is actually passed forward after attention weighting.


```python
# ==============================================================================
# DECODER Q/K/V AUDIT FOR THE SAME LOCAL INFORMATION UNITS
# ==============================================================================


def normalize_token_share(values):
    """Non-negative token magnitudes -> shares that sum to one within an event."""
    values = values.clamp(min=0)
    return values / (values.sum(dim=-1, keepdim=True) + ROUTING_EPS)


def encoded_memory_for_qkv(params, tokens):
    """Run the trained encoder exactly as in the model and return normalized memory."""
    for block in params["encoder_blocks"]:
        normalized = block["norm_attention"](tokens)
        attention_output = block["attention"](
            normalized, normalized, normalized, need_weights=False
        )[0]
        tokens = tokens + attention_output

        normalized = block["norm_ffn"](tokens)
        tokens = tokens + block["ffn_out"](F.gelu(block["ffn_in"](normalized)))

    return params["encoder_norm_out"](tokens)


def decoder_qkv_feature_shares(params, batch):
    """Q/K/V mechanism at the decoder, returned as feature-token shares.

    Q-K selection is the actual head-averaged softmax(QK'/sqrt(d_h)).
    V payload is the norm of each feature's multi-head V representation.
    Delivered V is the norm of each feature's attention-weighted V contribution
    after the trained decoder output projection (bias excluded because it is shared).
    """
    with torch.no_grad():
        tokens = build_tokens(params, batch)
        memory = encoded_memory_for_qkv(params, tokens)
        query = memory[:, :1, :]
        feature_memory = memory[:, 1:, :]

        layer = params["decoder_attention"]
        d_model = layer.embed_dim
        n_heads = layer.num_heads
        head_width = d_model // n_heads

        weight_q, weight_k, weight_v = layer.in_proj_weight.chunk(3, dim=0)
        bias_q, bias_k, bias_v = layer.in_proj_bias.chunk(3, dim=0)

        def split_heads(values):
            rows, length, _ = values.shape
            return values.view(rows, length, n_heads, head_width).transpose(1, 2)

        q = split_heads(F.linear(query, weight_q, bias_q))
        k = split_heads(F.linear(feature_memory, weight_k, bias_k))
        v = split_heads(F.linear(feature_memory, weight_v, bias_v))

        logits = q @ k.transpose(-2, -1) / math.sqrt(head_width)
        attention = torch.softmax(logits, dim=-1)                    # rows, heads, 1, features

        # Q-K: actual decoder selection probability, averaged over heads.
        qk_selection = attention.mean(dim=1)[:, 0, :]                # rows, features; sums to 1

        # V: size of the content made available by each feature before selection.
        v_concat = v.transpose(1, 2).reshape(v.shape[0], v.shape[2], d_model)
        v_payload = normalize_token_share(v_concat.norm(dim=-1))

        # Feature-specific A*V contribution, concatenated over heads and passed through W_o.
        weighted_v = attention[:, :, 0, :, None] * v                 # rows, heads, features, head_width
        weighted_concat = weighted_v.permute(0, 2, 1, 3).reshape(
            v.shape[0], v.shape[2], d_model
        )
        delivered_vectors = F.linear(weighted_concat, layer.out_proj.weight, bias=None)
        delivered_value = normalize_token_share(delivered_vectors.norm(dim=-1))

    return {
        "qk_selection": qk_selection,
        "v_payload": v_payload,
        "delivered_value": delivered_value,
    }


def qkv_audit_for_event(params, view, position, local_result):
    """Attach decoder Q/K/V mechanism shares to the matched-intervention summary."""
    batch = select_rows_from_split(view, slice(position, position + 1))
    feature_shares = decoder_qkv_feature_shares(params, batch)

    qk_units = aggregate_routing_by_unit(feature_shares["qk_selection"][0])
    v_units = aggregate_routing_by_unit(feature_shares["v_payload"][0])
    delivered_units = aggregate_routing_by_unit(feature_shares["delivered_value"][0])

    summary = local_result["summary"].copy()
    summary["qk_selection"] = pd.Series(qk_units)
    summary["v_payload"] = pd.Series(v_units)
    summary["delivered_value"] = pd.Series(delivered_units)
    summary["robust_direction"] = (
        ((summary["p10 impact"] > 0) & (summary["p90 impact"] > 0))
        | ((summary["p10 impact"] < 0) & (summary["p90 impact"] < 0))
    )

    return {
        "position": position,
        "prediction": local_result["prediction"],
        "actual": local_result["actual"],
        "summary": summary,
    }


def plot_qkv_audit(result, frame, top=TOP_INTERVENTIONS_PER_EVENT):
    """Readable Q/K/V display with the same structure used for routing."""
    plot_share_panels(
        result,
        frame,
        value_columns=["qk_selection", "v_payload", "delivered_value"],
        panel_titles=["Q-K selection", "V payload", "Delivered V"],
        main_title="decoder Q/K/V mechanism",
        top=top,
    )

# Consistency check: Q-K selection must reproduce the ordinary decoder attention share.
qkv_check_batch = select_rows_from_split(
    explain_views[EXPLAIN_SPLIT], slice(0, min(64, len(event_frame)))
)
with torch.no_grad():
    qkv_check = decoder_qkv_feature_shares(explain_params, qkv_check_batch)["qk_selection"]
    _, _, decoder_check_map = forward_with_routing_maps(
        explain_params,
        build_tokens(explain_params, qkv_check_batch),
        qkv_check_batch["exposure"],
    )
    decoder_check = decoder_attention_share(decoder_check_map)
qkv_gap = float((qkv_check - decoder_check).abs().max())
print(f"Largest Q-K selection vs decoder-attention difference: {qkv_gap:.2e}")
print("OK" if qkv_gap < 1e-6 else "CHECK THE Q/K PROJECTION")



def run_qkv_cases(case_kind, positions_group):
    """Decoder Q/K/V chart for every event of a single case group."""
    print(f"===== {case_kind.upper()} CASES: Q/K/V =====")
    for position in positions_group:
        position = int(position)
        QKV_AUDIT[position] = qkv_audit_for_event(
            explain_params,
            explain_views[EXPLAIN_SPLIT],
            position,
            MATCHED_LOCAL[position],
        )
        plot_qkv_audit(QKV_AUDIT[position], event_frame)


QKV_AUDIT = {}
```

### Step 8: Global dependency check

**Question:** does the fitted model depend on the same information across a whole out-of-sample year?  
Positive permutation effects mean performance deteriorates after coherent shuffling. MPD and D² stay separate.


```python
# ==============================================================================
# PERMUTATION IMPORTANCE: COHERENT UNITS, MPD + D2
# ==============================================================================

def poisson_d2(observed, predicted):
    observed = np.asarray(observed, dtype=float)
    predicted = np.asarray(predicted, dtype=float)
    if observed.sum() == 0:
        return np.nan
    return d2_tweedie_score(observed, predicted, power=1)


def permutation_importance(params, view, repeats=PERMUTATION_REPEATS, seed=EXPLAIN_SEED):
    contents = []
    n_rows = len(view["y"])
    for start in range(0, n_rows, PREDICT_BATCH_SIZE):
        stop = min(start + PREDICT_BATCH_SIZE, n_rows)
        batch = select_rows_from_split(view, slice(start, stop))
        contents.append(token_content(params, batch))

    content = torch.cat(contents, dim=0)
    exposure = view["exposure"]
    observed = view["y"][:, 0].cpu().numpy()
    full_prediction = expected_from_content(params, content, exposure).cpu().numpy()
    reference_mpd = mean_poisson_deviance(observed, full_prediction)
    reference_d2 = poisson_d2(observed, full_prediction)

    generator = torch.Generator(device="cpu").manual_seed(seed)
    rows = []

    for unit_name, token_names in INTERVENTION_UNITS.items():
        mpd_increases = []
        d2_decreases = []
        for _ in range(repeats):
            order = torch.randperm(len(content), generator=generator).to(content.device)
            shuffled = content.clone()
            for token_name in token_names:
                token_position = TOKEN_POSITION[token_name]
                shuffled[:, token_position] = content[order, token_position]

            shuffled_prediction = expected_from_content(params, shuffled, exposure).cpu().numpy()
            shuffled_mpd = mean_poisson_deviance(observed, shuffled_prediction)
            shuffled_d2 = poisson_d2(observed, shuffled_prediction)
            mpd_increases.append(shuffled_mpd - reference_mpd)
            d2_decreases.append(reference_d2 - shuffled_d2)

        rows.append({
            "unit": unit_name,
            "members": ", ".join(token_names),
            "MPD increase": float(np.mean(mpd_increases)),
            "MPD spread": float(np.std(mpd_increases, ddof=1)),
            "D2 decrease": float(np.mean(d2_decreases)),
            "D2 spread": float(np.std(d2_decreases, ddof=1)),
        })

    baseline = {"MPD": float(reference_mpd), "D2": float(reference_d2)}
    return pd.DataFrame(rows).set_index("unit"), baseline


PERMUTATION = {}
PERMUTATION_BASELINE = {}
for split_name in PERMUTATION_SPLITS:
    PERMUTATION[split_name], PERMUTATION_BASELINE[split_name] = permutation_importance(
        explain_params, explain_views[split_name]
    )
    year = splits_raw[split_name].index.year.min()
    baseline = PERMUTATION_BASELINE[split_name]
    print(f"{year}: all features -> MPD {baseline['MPD']:.4f} | D2 {baseline['D2']:.4f}")
```

### Global charts

A bar to the right of zero means shuffling worsens the metric; a bar to the left means the learned relationship was not helping that year.


```python
# ==============================================================================
# GLOBAL RESULT
# ==============================================================================

def plot_permutation_metric(permutation, value_column, spread_column, x_label, title, top=15):
    total = sum(frame[value_column] for frame in permutation.values())
    order = total.sort_values(ascending=False).index[:top][::-1]

    figure, axis = plt.subplots(figsize=(11, 0.50 * top + 1.8))
    positions = np.arange(len(order))
    bar_height = 0.8 / len(permutation)

    for offset, (split_name, frame) in enumerate(permutation.items()):
        year = splits_raw[split_name].index.year.min()
        shift = (offset - (len(permutation) - 1) / 2) * bar_height
        axis.barh(
            positions + shift,
            frame.loc[order, value_column],
            xerr=frame.loc[order, spread_column],
            height=bar_height * 0.9,
            label=str(year),
            error_kw={"elinewidth": 1, "ecolor": "#52514e"},
        )

    axis.axvline(0, color=COLOR_NEUTRAL, linewidth=1)
    axis.set_yticks(positions, [intervention_label(unit) for unit in order])
    axis.set_xlabel(x_label)
    axis.set_title(title, loc="left")
    axis.legend(title="out-of-sample year", frameon=False)
    axis.spines[["top", "right"]].set_visible(False)
    axis.grid(axis="x", alpha=0.3)
    plt.tight_layout()
    plt.show()





```
